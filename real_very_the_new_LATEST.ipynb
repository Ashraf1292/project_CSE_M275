```python
import pandas as pd
import numpy as np

# Number of per-class metrics actually returned by Ultralytics
n_metrics = len(results.box.ap50)

rows = []

for class_id in range(n_metrics):
    rows.append({
        "class_id": class_id,
        "class_name": class_names[class_id],
        "train_count": train_counts[class_id],
        "precision": results.box.p[class_id],
        "recall": results.box.r[class_id],
        "mAP50": results.box.ap50[class_id],
        "mAP50_95": results.box.ap[class_id]
    })

df = pd.DataFrame(rows)

print("Number of classes included:", len(df))
display(df.head(15))
```

```python
from scipy.stats import spearmanr, pearsonr

metrics = ["precision", "recall", "mAP50", "mAP50_95"]

print("===== CORRELATION WITH TRAINING FREQUENCY =====")

for metric in metrics:

    # Spearman correlation
    rho, p_value = spearmanr(
        df["train_count"],
        df[metric]
    )

    print(
        f"{metric:12s} | "
        f"Spearman rho = {rho: .4f} | "
        f"p-value = {p_value:.6f}"
    )
```

```python
from pathlib import Path

train_label_dir = Path("/content/BLPD/LPNR_Dataset/train/labels")

label_files = list(train_label_dir.glob("*.txt"))

total_objects = 0
empty_files = 0
malformed_lines = 0

for label_file in label_files:
    with open(label_file, "r") as f:
        lines = [line.strip() for line in f if line.strip()]

    if len(lines) == 0:
        empty_files += 1

    for line in lines:
        parts = line.split()

        if len(parts) >= 5:
            total_objects += 1
        else:
            malformed_lines += 1

print("Label files:", len(label_files))
print("Total objects:", total_objects)
print("Empty label files:", empty_files)
print("Malformed annotation lines:", malformed_lines)
```

```python
from collections import Counter

class_counter = Counter()

for label_file in label_files:
    with open(label_file, "r") as f:
        for line in f:
            parts = line.strip().split()

            if len(parts) >= 5:
                class_id = int(parts[0])
                class_counter[class_id] += 1

print("Number of classes found:", len(class_counter))
print("Total objects:", sum(class_counter.values()))

print("\nFirst 20 class counts:")
for class_id, count in sorted(class_counter.items())[:20]:
    print(class_id, class_names[class_id], count)
```

```python
from collections import Counter
from pathlib import Path

test_label_dir = Path("/content/BLPD/LPNR_Dataset/test/labels")

test_counts = Counter()

for label_file in test_label_dir.glob("*.txt"):
    with open(label_file, "r") as f:
        for line in f:
            parts = line.strip().split()

            if len(parts) >= 5:
                class_id = int(parts[0])
                test_counts[class_id] += 1

df["test_count"] = df["class_id"].map(test_counts).fillna(0).astype(int)

display(
    df.sort_values("train_count").head(20)
)
```

```python
print("===== TEST SUPPORT =====")

print("Total test objects:", df["test_count"].sum())
print("Classes with zero test instances:",
      (df["test_count"] == 0).sum())

display(
    df.sort_values("test_count").head(20)
)
```

```python
from scipy.stats import spearmanr

print("===== FREQUENCY vs PERFORMANCE =====")

for min_test_support in [2, 5, 10]:

    subset = df[df["test_count"] >= min_test_support].copy()

    print(f"\n--- Classes with test_count >= {min_test_support} ---")
    print("Number of classes:", len(subset))

    for metric in ["precision", "recall", "mAP50", "mAP50_95"]:

        rho, p_value = spearmanr(
            subset["train_count"],
            subset[metric]
        )

        print(
            f"{metric:12s} | "
            f"rho = {rho: .4f} | "
            f"p = {p_value:.6f}"
        )
```

```python
df["localization_gap"] = df["mAP50"] - df["mAP50_95"]

print("===== LARGEST mAP50 → mAP50-95 GAPS =====")

display(
    df.sort_values("localization_gap", ascending=False)[
        [
            "class_id",
            "class_name",
            "train_count",
            "test_count",
            "mAP50",
            "mAP50_95",
            "localization_gap"
        ]
    ].head(20)
)
```

```python
from scipy.stats import spearmanr

subset = df[df["test_count"] >= 5].copy()

rho, p = spearmanr(
    subset["train_count"],
    subset["localization_gap"]
)

print("Training frequency vs mAP50-mAP50_95 gap")
print("Spearman rho:", round(rho, 4))
print("p-value:", round(p, 6))
```

```python
print("===== LOWEST mAP50-95, TEST SUPPORT >= 10 =====")

display(
    df[df["test_count"] >= 10]
    .sort_values("mAP50_95")[
        [
            "class_id",
            "class_name",
            "train_count",
            "test_count",
            "precision",
            "recall",
            "mAP50",
            "mAP50_95",
            "localization_gap"
        ]
    ]
    .head(20)
)
```

```python
from collections import Counter
import os
import yaml

# Paths
label_dir = "/content/BLPD/LPNR_Dataset/train/labels"
yaml_path = "/content/BLPD/LPNR_data.yaml"

# Load class names
with open(yaml_path, "r") as f:
    data = yaml.safe_load(f)

names = data["names"]

# Count valid detection annotations
class_counts = Counter()

for filename in os.listdir(label_dir):
    if not filename.endswith(".txt"):
        continue

    with open(os.path.join(label_dir, filename), "r") as f:
        for line in f:
            parts = line.strip().split()

            # Only count valid YOLO detection rows
            if len(parts) == 5:
                class_id = int(parts[0])
                class_counts[class_id] += 1

print("Total valid annotations:", sum(class_counts.values()))
print("Classes represented:", len(class_counts))

print("\n10 most frequent classes:")
for class_id, count in class_counts.most_common(10):
    print(f"{class_id:3d} | {names[class_id]:15s} | {count}")

print("\n10 least frequent classes:")
for class_id, count in sorted(class_counts.items(), key=lambda x: x[1])[:10]:
    print(f"{class_id:3d} | {names[class_id]:15s} | {count}")
```

```python
from pathlib import Path
from PIL import Image
import pandas as pd
import numpy as np

val_images = Path("/content/BLPD/LPNR_Dataset/valid/images")
val_labels = Path("/content/BLPD/LPNR_Dataset/valid/labels")

records = []

for label_file in sorted(val_labels.glob("*.txt")):

    # Find corresponding image
    image_file = None
    for ext in [".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG"]:
        candidate = val_images / (label_file.stem + ext)
        if candidate.exists():
            image_file = candidate
            break

    if image_file is None:
        continue

    # Actual image dimensions
    with Image.open(image_file) as img:
        img_w, img_h = img.size

    # Read annotations
    with open(label_file, "r") as f:
        lines = [line.strip() for line in f if line.strip()]

    for line in lines:
        parts = line.split()

        if len(parts) < 5:
            continue

        class_id = int(parts[0])
        x_center = float(parts[1])
        y_center = float(parts[2])
        box_w_norm = float(parts[3])
        box_h_norm = float(parts[4])

        # Convert normalized dimensions to pixels
        box_w_px = box_w_norm * img_w
        box_h_px = box_h_norm * img_h

        area_px = box_w_px * box_h_px
        area_fraction = box_w_norm * box_h_norm

        records.append({
            "image": image_file.name,
            "class_id": class_id,
            "img_w": img_w,
            "img_h": img_h,
            "box_w_px": box_w_px,
            "box_h_px": box_h_px,
            "box_area_px": area_px,
            "box_area_fraction": area_fraction,
            "min_dimension_px": min(box_w_px, box_h_px)
        })

df = pd.DataFrame(records)

print("Total annotated objects:", len(df))
print("Total validation images:", df["image"].nunique())

print("\nImage dimensions:")
print(df[["img_w", "img_h"]].drop_duplicates().to_string(index=False))

print("\nObject-size summary:")
print(
    df[
        ["box_w_px", "box_h_px", "box_area_px",
         "box_area_fraction", "min_dimension_px"]
    ].describe(
        percentiles=[0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    ).round(3)
)
```

```python
from pathlib import Path
from ultralytics import YOLO
import pandas as pd
import numpy as np
import gc
import torch

# --------------------------------------------------
# Clean up CUDA
# --------------------------------------------------

gc.collect()
torch.cuda.empty_cache()

# --------------------------------------------------
# Load model
# --------------------------------------------------

try:
    model
    print("Using existing model.")
except NameError:
    model = YOLO("/content/BLPD/LPNR_Model.pt")
    print("Loaded LPNR_Model.pt.")

# --------------------------------------------------
# Paths
# --------------------------------------------------

val_images = Path("/content/BLPD/LPNR_Dataset/valid/images")
val_labels = Path("/content/BLPD/LPNR_Dataset/valid/labels")

image_files = sorted(val_images.glob("*.jpg"))

print(f"Validation images: {len(image_files)}")


# --------------------------------------------------
# IoU helper
# --------------------------------------------------

def box_iou(box1, boxes2):

    if len(boxes2) == 0:
        return np.array([])

    x1 = np.maximum(box1[0], boxes2[:, 0])
    y1 = np.maximum(box1[1], boxes2[:, 1])
    x2 = np.minimum(box1[2], boxes2[:, 2])
    y2 = np.minimum(box1[3], boxes2[:, 3])

    inter_w = np.maximum(0, x2 - x1)
    inter_h = np.maximum(0, y2 - y1)
    inter = inter_w * inter_h

    area1 = max(0, box1[2] - box1[0]) * \
            max(0, box1[3] - box1[1])

    area2 = np.maximum(0, boxes2[:, 2] - boxes2[:, 0]) * \
            np.maximum(0, boxes2[:, 3] - boxes2[:, 1])

    union = area1 + area2 - inter

    return np.where(union > 0, inter / union, 0)


# --------------------------------------------------
# Process images ONE AT A TIME
# --------------------------------------------------

all_results = []

for idx, image_path in enumerate(image_files):

    label_path = val_labels / f"{image_path.stem}.txt"

    if not label_path.exists():
        continue

    # ---- inference on ONE image ----
    result = model.predict(
        source=str(image_path),
        imgsz=640,
        conf=0.001,
        iou=0.7,
        max_det=300,
        save=False,
        verbose=False
    )[0]

    # ---- predictions ----
    pred_boxes = result.boxes.xyxy.cpu().numpy()
    pred_classes = result.boxes.cls.cpu().numpy().astype(int)
    pred_conf = result.boxes.conf.cpu().numpy()

    img_h, img_w = result.orig_shape

    # ---- ground truth ----
    with open(label_path, "r") as f:
        lines = [line.strip() for line in f if line.strip()]

    for line in lines:

        parts = line.split()

        if len(parts) < 5:
            continue

        true_class = int(parts[0])

        xc = float(parts[1]) * img_w
        yc = float(parts[2]) * img_h
        w = float(parts[3]) * img_w
        h = float(parts[4]) * img_h

        gt_box = [
            xc - w / 2,
            yc - h / 2,
            xc + w / 2,
            yc + h / 2
        ]

        # ---- best overlap ----
        ious = box_iou(gt_box, pred_boxes)

        if len(ious) == 0:
            best_iou = 0.0
            best_class = -1
            best_conf = 0.0
        else:
            best_idx = int(np.argmax(ious))
            best_iou = float(ious[best_idx])
            best_class = int(pred_classes[best_idx])
            best_conf = float(pred_conf[best_idx])

        # ---- object size ----
        min_dim = min(w, h)
        area = w * h

        # ---- indicators ----
        all_results.append({
            "image": image_path.name,
            "true_class": true_class,

            "box_w_px": w,
            "box_h_px": h,
            "min_dimension_px": min_dim,
            "box_area_px": area,

            "best_iou": best_iou,
            "best_pred_class": best_class,
            "best_conf": best_conf,

            "detected_iou50": best_iou >= 0.50,
            "detected_iou75": best_iou >= 0.75,

            "correct_class": (
                best_class == true_class
                if best_class >= 0 else False
            ),

            "correct_class_iou50": (
                best_iou >= 0.50 and
                best_class == true_class
            ),

            "correct_class_iou75": (
                best_iou >= 0.75 and
                best_class == true_class
            )
        })

    # Progress
    if (idx + 1) % 100 == 0:
        print(f"Processed {idx + 1}/{len(image_files)} images")

    # Periodic cleanup
    if (idx + 1) % 200 == 0:
        gc.collect()
        torch.cuda.empty_cache()


# --------------------------------------------------
# Build dataframe
# --------------------------------------------------

match_df = pd.DataFrame(all_results)

print("\n======================================")
print("ANALYSIS COMPLETE")
print("======================================")

print("Ground-truth objects analysed:", len(match_df))

print("\nOverall diagnostic statistics:")
print(
    match_df[
        [
            "detected_iou50",
            "detected_iou75",
            "correct_class_iou50",
            "correct_class_iou75"
        ]
    ].mean().round(4)
)

# --------------------------------------------------
# Save
# --------------------------------------------------

output_path = "/content/BLPD/LPNR_validation_object_analysis.csv"

match_df.to_csv(output_path, index=False)

print("\nSaved to:")
print(output_path)
```

```python
from pathlib import Path
import cv2
import pandas as pd
import numpy as np

# Load our previous object-level analysis
df = pd.read_csv("/content/BLPD/LPNR_validation_object_analysis.csv")

val_images = Path("/content/BLPD/LPNR_Dataset/valid/images")

# Calculate image sharpness for each validation image
sharpness = {}

image_files = list(val_images.glob("*"))

for i, image_path in enumerate(image_files):
    img = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)

    if img is None:
        continue

    laplacian = cv2.Laplacian(img, cv2.CV_64F)
    sharpness[image_path.name] = laplacian.var()

    if (i + 1) % 300 == 0:
        print(f"Processed {i+1}/{len(image_files)} images")

# Add image-level sharpness to every character/object
df["sharpness"] = df["image"].map(sharpness)

print("\nImages with sharpness values:",
      df["sharpness"].notna().sum())

print("\nSharpness statistics:")
print(df["sharpness"].describe().round(2))
```

```python
# Create sharpness quartile groups
df["sharpness_group"] = pd.qcut(
    df["sharpness"],
    q=4,
    labels=[
        "Lowest sharpness",
        "Low-medium sharpness",
        "Medium-high sharpness",
        "Highest sharpness"
    ],
    duplicates="drop"
)

sharpness_summary = (
    df.groupby("sharpness_group", observed=False)
      .agg(
          objects=("true_class", "size"),
          mean_sharpness=("sharpness", "mean"),
          iou50_rate=("detected_iou50", "mean"),
          iou75_rate=("detected_iou75", "mean"),
          correct_iou50=("correct_class_iou50", "mean"),
          correct_iou75=("correct_class_iou75", "mean")
      )
      .reset_index()
)

for col in [
    "iou50_rate",
    "iou75_rate",
    "correct_iou50",
    "correct_iou75"
]:
    sharpness_summary[col] *= 100

print(sharpness_summary.round(2).to_string(index=False))
```

```python
# Create small-vs-normal groups
df["size_category"] = np.where(
    df["min_dimension_px"] < 32,
    "<32 px",
    ">=32 px"
)

class_size_summary = (
    df.groupby(["true_class", "size_category"])
      .agg(
          objects=("true_class", "size"),
          accuracy=("correct_class_iou50", "mean")
      )
      .reset_index()
)

# Convert accuracy to percentage
class_size_summary["accuracy"] *= 100

# Pivot so small and normal sizes are side by side
pivot = class_size_summary.pivot(
    index="true_class",
    columns="size_category",
    values=["objects", "accuracy"]
)

# Flatten column names
pivot.columns = [
    f"{metric}_{size.replace(' ', '_')}"
    for metric, size in pivot.columns
]

pivot = pivot.reset_index()

# Only retain classes having at least 5 small objects
filtered = pivot[
    pivot.get("objects_<32_px", pd.Series(0, index=pivot.index)) >= 5
].copy()

# Calculate degradation from >=32 px to <32 px
filtered["accuracy_drop"] = (
    filtered["accuracy_>=32_px"] -
    filtered["accuracy_<32_px"]
)

# Sort by largest degradation
filtered = filtered.sort_values(
    "accuracy_drop",
    ascending=False
)

print("Classes with at least 5 objects below 32 px:")
print(
    filtered[
        [
            "true_class",
            "objects_<32_px",
            "objects_>=32_px",
            "accuracy_<32_px",
            "accuracy_>=32_px",
            "accuracy_drop"
        ]
    ].round(2).to_string(index=False)
)
```

```python
from pathlib import Path
import cv2
import pandas as pd
import numpy as np
import math
from PIL import Image, ImageDraw, ImageFont

# Load the failure list
failures = pd.read_csv(
    "/content/BLPD/small_character_failures.csv"
).sort_values("min_dimension_px")

val_images = Path("/content/BLPD/LPNR_Dataset/valid/images")
val_labels = Path("/content/BLPD/LPNR_Dataset/valid/labels")

# Take the 30 smallest failures
samples = failures.head(30).copy()

# Output directory
output_dir = Path("/content/BLPD/small_character_visual_analysis")
output_dir.mkdir(exist_ok=True)

# Font
try:
    font = ImageFont.truetype(
        "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", 16
    )
except:
    font = ImageFont.load_default()

tiles = []

for idx, row in samples.iterrows():

    image_path = val_images / row["image"]
    label_path = val_labels / f'{Path(row["image"]).stem}.txt'

    img = cv2.imread(str(image_path))

    if img is None:
        continue

    h, w = img.shape[:2]

    # Find the matching GT box
    target_class = int(row["true_class"])

    best_match = None
    best_distance = float("inf")

    with open(label_path, "r") as f:
        for line in f:
            parts = line.strip().split()

            if len(parts) < 5:
                continue

            cls = int(parts[0])

            if cls != target_class:
                continue

            xc = float(parts[1]) * w
            yc = float(parts[2]) * h
            bw = float(parts[3]) * w
            bh = float(parts[4]) * h

            x1 = xc - bw / 2
            y1 = yc - bh / 2
            x2 = xc + bw / 2
            y2 = yc + bh / 2

            min_dim = min(bw, bh)

            distance = abs(min_dim - row["min_dimension_px"])

            if distance < best_distance:
                best_distance = distance
                best_match = (x1, y1, x2, y2)

    if best_match is None:
        continue

    x1, y1, x2, y2 = best_match

    # Add generous context around the tiny character
    cx = (x1 + x2) / 2
    cy = (y1 + y2) / 2

    crop_size = max(80, int(max(x2-x1, y2-y1) * 4))

    crop_x1 = max(0, int(cx - crop_size / 2))
    crop_y1 = max(0, int(cy - crop_size / 2))
    crop_x2 = min(w, int(cx + crop_size / 2))
    crop_y2 = min(h, int(cy + crop_size / 2))

    crop = img[crop_y1:crop_y2, crop_x1:crop_x2]

    if crop.size == 0:
        continue

    # Upscale for visual inspection
    crop = cv2.resize(
        crop,
        (240, 240),
        interpolation=cv2.INTER_NEAREST
    )

    crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)

    tile = Image.new("RGB", (300, 295), "white")
    tile.paste(Image.fromarray(crop_rgb), (30, 5))

    draw = ImageDraw.Draw(tile)

    text = (
        f"True: {int(row['true_class'])}  "
        f"Pred: {int(row['best_pred_class'])}\n"
        f"Size: {row['min_dimension_px']:.1f}px\n"
        f"IoU: {row['best_iou']:.3f}  "
        f"Conf: {row['best_conf']:.3f}"
    )

    draw.multiline_text(
        (10, 250),
        text,
        fill="black",
        font=font,
        spacing=2
    )

    tiles.append(tile)

# Create contact sheet
cols = 5
rows = math.ceil(len(tiles) / cols)

sheet = Image.new(
    "RGB",
    (cols * 300, rows * 295),
    "white"
)

for i, tile in enumerate(tiles):
    x = (i % cols) * 300
    y = (i // cols) * 295
    sheet.paste(tile, (x, y))

output_path = output_dir / "smallest_30_failures_contact_sheet.png"
sheet.save(output_path)

print("Created contact sheet:")
print(output_path)
print(f"Tiles included: {len(tiles)}")
```

```python
from pathlib import Path
from ultralytics import YOLO

base = Path("/content/BLPD/LPNR_Dataset")

splits = {
    "train": base / "train",
    "valid": base / "valid",
    "test": base / "test"
}

# Load the released LPNR checkpoint just to obtain class names
model = YOLO("/content/BLPD/LPNR_Model.pt")
names = model.names

print(f"Loaded {len(names)} classes.")


def get_plate_strings(split_dir):
    label_dir = split_dir / "labels"
    image_dir = split_dir / "images"

    plates = {}

    for label_file in label_dir.glob("*.txt"):

        image_file = None

        for ext in [".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG"]:
            candidate = image_dir / (label_file.stem + ext)
            if candidate.exists():
                image_file = candidate
                break

        if image_file is None:
            continue

        objects = []

        with open(label_file, "r") as f:
            for line in f:
                parts = line.strip().split()

                if len(parts) < 5:
                    continue

                cls = int(parts[0])
                x_center = float(parts[1])

                objects.append((x_center, cls))

        # Left-to-right character order
        objects.sort(key=lambda x: x[0])

        plate = " ".join(names[cls] for _, cls in objects)

        plates[label_file.stem] = plate

    return plates


# Process all splits
all_plates = {}

for split, split_dir in splits.items():
    print(f"\nProcessing {split}...")
    all_plates[split] = get_plate_strings(split_dir)
    print(f"Images processed: {len(all_plates[split])}")


# Unique plate strings
train_plates = set(all_plates["train"].values())
valid_plates = set(all_plates["valid"].values())
test_plates  = set(all_plates["test"].values())


print("\n" + "="*60)
print("UNIQUE PLATE STRING COUNTS")
print("="*60)

print("Train:", len(train_plates))
print("Valid:", len(valid_plates))
print("Test :", len(test_plates))


# Overlaps
train_valid = train_plates & valid_plates
train_test  = train_plates & test_plates
valid_test  = valid_plates & test_plates
all_three   = train_plates & valid_plates & test_plates


print("\n" + "="*60)
print("EXACT PLATE STRING OVERLAP")
print("="*60)

print("Train ∩ Valid:", len(train_valid))
print("Train ∩ Test :", len(train_test))
print("Valid ∩ Test :", len(valid_test))
print("All three    :", len(all_three))


print("\nOverlap percentages:")

print(
    f"Train → Valid: "
    f"{len(train_valid)/len(valid_plates)*100:.2f}%"
)

print(
    f"Train → Test:  "
    f"{len(train_test)/len(test_plates)*100:.2f}%"
)

print(
    f"Valid → Test:  "
    f"{len(valid_test)/len(test_plates)*100:.2f}%"
)


# Show examples
print("\n" + "="*60)
print("EXAMPLE TRAIN–TEST OVERLAPS")
print("="*60)

for plate in sorted(train_test)[:30]:
    print(plate)
```

```python
from pathlib import Path
from PIL import Image
import numpy as np
import cv2
import hashlib
from collections import defaultdict

base = Path("/content/BLPD/LPNR_Dataset")

# ---------------------------------------------------------
# Build image paths for every plate
# ---------------------------------------------------------

plate_paths = {
    "train": defaultdict(list),
    "test": defaultdict(list)
}

for split in ["train", "test"]:
    image_dir = base / split / "images"

    for image_id, plate in all_plates[split].items():

        image_path = None

        for ext in [".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG"]:
            candidate = image_dir / (image_id + ext)
            if candidate.exists():
                image_path = candidate
                break

        if image_path is not None:
            plate_paths[split][plate].append(image_path)


# ---------------------------------------------------------
# Simple perceptual similarity using normalized grayscale
# ---------------------------------------------------------

def image_similarity(path1, path2):
    img1 = cv2.imread(str(path1), cv2.IMREAD_GRAYSCALE)
    img2 = cv2.imread(str(path2), cv2.IMREAD_GRAYSCALE)

    if img1 is None or img2 is None:
        return None

    # Resize to the same small representation
    img1 = cv2.resize(img1, (128, 128))
    img2 = cv2.resize(img2, (128, 128))

    # Normalize brightness/contrast
    img1 = cv2.equalizeHist(img1)
    img2 = cv2.equalizeHist(img2)

    # Correlation
    correlation = np.corrcoef(
        img1.flatten(),
        img2.flatten()
    )[0, 1]

    return float(correlation)


# ---------------------------------------------------------
# Compare train/test images for overlapping plate strings
# ---------------------------------------------------------

overlap = set(plate_paths["train"]) & set(plate_paths["test"])

results = []

for i, plate in enumerate(sorted(overlap), 1):

    train_imgs = plate_paths["train"][plate]
    test_imgs = plate_paths["test"][plate]

    for train_img in train_imgs:
        for test_img in test_imgs:

            similarity = image_similarity(
                train_img,
                test_img
            )

            if similarity is not None:
                results.append({
                    "plate": plate,
                    "train_image": train_img.name,
                    "test_image": test_img.name,
                    "similarity": similarity
                })

    if i % 50 == 0:
        print(f"Processed {i}/{len(overlap)} plate strings")


# ---------------------------------------------------------
# Results
# ---------------------------------------------------------

print("\n" + "="*70)
print("IMAGE-LEVEL SIMILARITY RESULTS")
print("="*70)

print("Compared image pairs:", len(results))

similarities = np.array([
    r["similarity"] for r in results
])

print(f"Mean similarity  : {similarities.mean():.4f}")
print(f"Median similarity: {np.median(similarities):.4f}")
print(f"Maximum similarity: {similarities.max():.4f}")

print("\nSimilarity thresholds:")

for threshold in [0.70, 0.80, 0.90, 0.95, 0.98]:
    count = np.sum(similarities >= threshold)
    percentage = count / len(similarities) * 100

    print(
        f">= {threshold:.2f}: "
        f"{count} pairs ({percentage:.2f}%)"
    )


# ---------------------------------------------------------
# Show most similar train/test pairs
# ---------------------------------------------------------

print("\n" + "="*70)
print("TOP 30 MOST SIMILAR TRAIN-TEST PAIRS")
print("="*70)

top_results = sorted(
    results,
    key=lambda x: x["similarity"],
    reverse=True
)[:30]

for r in top_results:
    print(
        f"\nSimilarity: {r['similarity']:.4f}"
        f"\nPlate: {r['plate']}"
        f"\nTrain: {r['train_image']}"
        f"\nTest : {r['test_image']}"
    )
```

```python
from pathlib import Path
import shutil

base = Path("/content/BLPD/LPNR_Dataset")

train_dir = base / "train" / "images"
test_dir = base / "test" / "images"
test_labels_dir = base / "test" / "labels"

clean_dir = Path("/content/BLPD/Clean_Test")
clean_images = clean_dir / "images"
clean_labels = clean_dir / "labels"

# Remove an old version if it exists
if clean_dir.exists():
    shutil.rmtree(clean_dir)

clean_images.mkdir(parents=True)
clean_labels.mkdir(parents=True)


# ---------------------------------------------------------
# Get source IDs appearing in training
# ---------------------------------------------------------

def get_source_id(filename):
    name = Path(filename).name

    if ".rf." in name:
        return name.split(".rf.")[0]

    return Path(name).stem


train_source_ids = set()

for image_path in train_dir.iterdir():

    if image_path.is_file():
        train_source_ids.add(
            get_source_id(image_path.name)
        )


# ---------------------------------------------------------
# Copy only source-independent test images
# ---------------------------------------------------------

total_test = 0
removed = 0
kept = 0

for image_path in test_dir.iterdir():

    if not image_path.is_file():
        continue

    total_test += 1

    source_id = get_source_id(image_path.name)

    # Exclude if the original source appears in training
    if source_id in train_source_ids:
        removed += 1
        continue

    # Keep image
    shutil.copy2(
        image_path,
        clean_images / image_path.name
    )

    # Copy corresponding label
    label_path = test_labels_dir / f"{image_path.stem}.txt"

    if label_path.exists():
        shutil.copy2(
            label_path,
            clean_labels / label_path.name
        )

    kept += 1


print("=" * 60)
print("CLEAN TEST SET CREATED")
print("=" * 60)

print("Original test images :", total_test)
print("Removed (train overlap):", removed)
print("Clean test images    :", kept)

print(
    "Remaining percentage :",
    f"{kept / total_test * 100:.2f}%"
)

print("\nLocation:")
print(clean_dir)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# Paths
# ---------------------------------------------------------
original_labels = Path("/content/BLPD/LPNR_Dataset/test/labels")
clean_labels = Path("/content/BLPD/Clean_Test/labels")

# YOLO labels are normalized; dataset images are 640x640
IMG_SIZE = 640

# Same bins we used previously
bins = [0, 16, 32, 48, 64, 96, 128, 192, float("inf")]
bin_labels = [
    "<16",
    "16–31",
    "32–47",
    "48–63",
    "64–95",
    "96–127",
    "128–191",
    "≥192"
]

def collect_min_dimensions(label_dir):
    values = []

    for label_file in sorted(label_dir.glob("*.txt")):
        with open(label_file, "r") as f:
            for line in f:
                parts = line.strip().split()

                # Standard YOLO detection format:
                # class x_center y_center width height
                if len(parts) < 5:
                    continue

                w = float(parts[3]) * IMG_SIZE
                h = float(parts[4]) * IMG_SIZE

                values.append(min(w, h))

    return np.array(values)


# ---------------------------------------------------------
# Collect object sizes
# ---------------------------------------------------------
original_sizes = collect_min_dimensions(original_labels)
clean_sizes = collect_min_dimensions(clean_labels)

print("Original test objects:", len(original_sizes))
print("Clean test objects:", len(clean_sizes))


# ---------------------------------------------------------
# Summary statistics
# ---------------------------------------------------------
def summarize(values, name):
    print(f"\n{name}")
    print("-" * 50)

    stats = {
        "Mean": np.mean(values),
        "Std": np.std(values),
        "Min": np.min(values),
        "10th percentile": np.percentile(values, 10),
        "25th percentile": np.percentile(values, 25),
        "Median": np.percentile(values, 50),
        "75th percentile": np.percentile(values, 75),
        "90th percentile": np.percentile(values, 90),
        "Max": np.max(values),
    }

    for k, v in stats.items():
        print(f"{k:18s}: {v:.2f} px")


summarize(original_sizes, "ORIGINAL TEST")
summarize(clean_sizes, "CLEAN TEST")


# ---------------------------------------------------------
# Size-bin comparison
# ---------------------------------------------------------
original_counts, _ = np.histogram(original_sizes, bins=bins)
clean_counts, _ = np.histogram(clean_sizes, bins=bins)

comparison = pd.DataFrame({
    "Size group": bin_labels,
    "Original test": original_counts,
    "Original %": original_counts / len(original_sizes) * 100,
    "Clean test": clean_counts,
    "Clean %": clean_counts / len(clean_sizes) * 100,
})

comparison["Difference (percentage points)"] = (
    comparison["Clean %"] - comparison["Original %"]
)

print("\nSIZE DISTRIBUTION")
print("=" * 90)

display(
    comparison.style.format({
        "Original %": "{:.2f}%",
        "Clean %": "{:.2f}%",
        "Difference (percentage points)": "{:+.2f}"
    })
)


# ---------------------------------------------------------
# Key small-object comparisons
# ---------------------------------------------------------
print("\nKEY COMPARISONS")
print("=" * 50)

for threshold in [16, 32, 48, 64, 96]:
    orig_pct = np.mean(original_sizes < threshold) * 100
    clean_pct = np.mean(clean_sizes < threshold) * 100

    print(
        f"< {threshold:3d}px: "
        f"Original = {orig_pct:6.2f}% | "
        f"Clean = {clean_pct:6.2f}% | "
        f"Difference = {clean_pct - orig_pct:+6.2f} pp"
    )
```

```python
# ============================================================
# CLEAN TEST — OBJECT-LEVEL FAILURE ANALYSIS
# CPU-SAFE, ONE IMAGE AT A TIME
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import cv2
import gc
import torch
from ultralytics import YOLO

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------
model_path = "/content/BLPD/LPNR_Model.pt"
image_dir = Path("/content/BLPD/Clean_Test/images")
label_dir = Path("/content/BLPD/Clean_Test/labels")

output_dir = Path("/content/BLPD/Clean_Test/object_analysis")
output_dir.mkdir(parents=True, exist_ok=True)

output_csv = output_dir / "clean_test_object_analysis.csv"

# ------------------------------------------------------------
# 2. Load model
# ------------------------------------------------------------
print("Loading model...")

model = YOLO(model_path)

print("Model loaded.")
print("Classes:", len(model.names))
print("Device: CPU")

# ------------------------------------------------------------
# 3. Get image list
# ------------------------------------------------------------
image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

image_paths = sorted([
    p for p in image_dir.iterdir()
    if p.suffix.lower() in image_extensions
])

print(f"\nClean-test images found: {len(image_paths)}")

# ------------------------------------------------------------
# 4. Helper: IoU
# ------------------------------------------------------------
def calculate_iou(box1, box2):
    """
    Boxes are [x1, y1, x2, y2]
    """
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    inter_w = max(0, x2 - x1)
    inter_h = max(0, y2 - y1)
    inter_area = inter_w * inter_h

    area1 = max(0, box1[2] - box1[0]) * max(0, box1[3] - box1[1])
    area2 = max(0, box2[2] - box2[0]) * max(0, box2[3] - box2[1])

    union = area1 + area2 - inter_area

    if union <= 0:
        return 0.0

    return inter_area / union


# ------------------------------------------------------------
# 5. Process images one at a time
# ------------------------------------------------------------
all_rows = []

for idx, image_path in enumerate(image_paths, start=1):

    # Corresponding label file
    label_path = label_dir / f"{image_path.stem}.txt"

    if not label_path.exists():
        print(f"WARNING: Missing label for {image_path.name}")
        continue

    # Read image
    image = cv2.imread(str(image_path))

    if image is None:
        print(f"WARNING: Could not read {image_path.name}")
        continue

    img_h, img_w = image.shape[:2]

    # --------------------------------------------------------
    # Ground-truth objects
    # --------------------------------------------------------
    gt_objects = []

    with open(label_path, "r") as f:
        for line in f:

            parts = line.strip().split()

            if len(parts) < 5:
                continue

            gt_class = int(float(parts[0]))

            xc = float(parts[1]) * img_w
            yc = float(parts[2]) * img_h
            w = float(parts[3]) * img_w
            h = float(parts[4]) * img_h

            x1 = max(0, xc - w / 2)
            y1 = max(0, yc - h / 2)
            x2 = min(img_w, xc + w / 2)
            y2 = min(img_h, yc + h / 2)

            gt_objects.append({
                "class_id": gt_class,
                "box": [x1, y1, x2, y2],
                "width": w,
                "height": h,
                "min_dimension": min(w, h),
            })

    if len(gt_objects) == 0:
        continue

    # --------------------------------------------------------
    # Run prediction — ONE IMAGE ONLY
    # --------------------------------------------------------
    results = model.predict(
        source=str(image_path),
        device="cpu",
        verbose=False,
        save=False
    )

    result = results[0]

    # --------------------------------------------------------
    # Extract predictions
    # --------------------------------------------------------
    predictions = []

    if result.boxes is not None and len(result.boxes) > 0:

        pred_boxes = result.boxes.xyxy.cpu().numpy()
        pred_classes = result.boxes.cls.cpu().numpy().astype(int)
        pred_conf = result.boxes.conf.cpu().numpy()

        for box, cls, conf in zip(
            pred_boxes,
            pred_classes,
            pred_conf
        ):
            predictions.append({
                "class_id": int(cls),
                "box": box.tolist(),
                "confidence": float(conf)
            })

    # --------------------------------------------------------
    # Match each GT object to its best-overlap prediction
    # --------------------------------------------------------
    for gt_idx, gt in enumerate(gt_objects):

        best_iou = 0.0
        best_prediction = None

        for pred in predictions:

            iou = calculate_iou(
                gt["box"],
                pred["box"]
            )

            if iou > best_iou:
                best_iou = iou
                best_prediction = pred

        # ----------------------------------------------------
        # Prediction information
        # ----------------------------------------------------
        if best_prediction is not None:

            pred_class = best_prediction["class_id"]
            confidence = best_prediction["confidence"]

        else:

            pred_class = -1
            confidence = 0.0

        correct_class = (
            best_prediction is not None
            and pred_class == gt["class_id"]
        )

        iou50 = best_iou >= 0.50
        iou75 = best_iou >= 0.75

        correct_iou50 = correct_class and iou50
        correct_iou75 = correct_class and iou75

        # ----------------------------------------------------
        # Size group
        # ----------------------------------------------------
        d = gt["min_dimension"]

        if d < 16:
            size_group = "<16"
        elif d < 32:
            size_group = "16–31"
        elif d < 48:
            size_group = "32–47"
        elif d < 64:
            size_group = "48–63"
        elif d < 96:
            size_group = "64–95"
        elif d < 128:
            size_group = "96–127"
        elif d < 192:
            size_group = "128–191"
        else:
            size_group = "≥192"

        # ----------------------------------------------------
        # Save row
        # ----------------------------------------------------
        all_rows.append({
            "image": image_path.name,
            "gt_index": gt_idx,

            "true_class": gt["class_id"],
            "pred_class": pred_class,

            "width_px": gt["width"],
            "height_px": gt["height"],
            "min_dimension_px": gt["min_dimension"],

            "size_group": size_group,

            "best_iou": best_iou,
            "confidence": confidence,

            "correct_class": int(correct_class),
            "iou50": int(iou50),
            "iou75": int(iou75),

            "correct_iou50": int(correct_iou50),
            "correct_iou75": int(correct_iou75),
        })

    # --------------------------------------------------------
    # Progress
    # --------------------------------------------------------
    if idx % 25 == 0 or idx == len(image_paths):

        print(
            f"Processed {idx}/{len(image_paths)} images "
            f"({idx / len(image_paths) * 100:.1f}%)"
        )

    # Release temporary objects
    del results
    del result
    del image

    gc.collect()


# ------------------------------------------------------------
# 6. Save results
# ------------------------------------------------------------
df = pd.DataFrame(all_rows)

df.to_csv(output_csv, index=False)

print("\n" + "=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)

print("Images processed:", len(image_paths))
print("Ground-truth objects analyzed:", len(df))

print("\nSaved to:")
print(output_csv)

# ------------------------------------------------------------
# 7. Basic summary
# ------------------------------------------------------------
print("\nOverall object-level results:")
print("-" * 50)

print(
    f"Correct class:       "
    f"{df['correct_class'].mean() * 100:.2f}%"
)

print(
    f"IoU ≥ 0.50:          "
    f"{df['iou50'].mean() * 100:.2f}%"
)

print(
    f"IoU ≥ 0.75:          "
    f"{df['iou75'].mean() * 100:.2f}%"
)

print(
    f"Correct + IoU ≥ 0.50:"
    f" {df['correct_iou50'].mean() * 100:.2f}%"
)

print(
    f"Correct + IoU ≥ 0.75:"
    f" {df['correct_iou75'].mean() * 100:.2f}%"
)

# ------------------------------------------------------------
# 8. Size-group performance
# ------------------------------------------------------------
size_order = [
    "<16",
    "16–31",
    "32–47",
    "48–63",
    "64–95",
    "96–127",
    "128–191",
    "≥192"
]

size_summary = (
    df.groupby("size_group", observed=False)
      .agg(
          objects=("true_class", "size"),
          mean_size_px=("min_dimension_px", "mean"),
          iou50_rate=("iou50", "mean"),
          iou75_rate=("iou75", "mean"),
          correct_class=("correct_class", "mean"),
          correct_iou50=("correct_iou50", "mean"),
          correct_iou75=("correct_iou75", "mean")
      )
      .reindex(size_order)
      .reset_index()
)

for col in [
    "iou50_rate",
    "iou75_rate",
    "correct_class",
    "correct_iou50",
    "correct_iou75"
]:
    size_summary[col] *= 100

print("\nSIZE-BASED PERFORMANCE")
print("=" * 90)

display(
    size_summary.style.format({
        "mean_size_px": "{:.2f}",
        "iou50_rate": "{:.2f}%",
        "iou75_rate": "{:.2f}%",
        "correct_class": "{:.2f}%",
        "correct_iou50": "{:.2f}%",
        "correct_iou75": "{:.2f}%"
    })
)

# Save summary too
size_summary.to_csv(
    output_dir / "clean_test_size_performance.csv",
    index=False
)

print("\nSize summary saved to:")
print(output_dir / "clean_test_size_performance.csv")
```

```python
# ============================================================
# PER-CHARACTER SIZE VULNERABILITY ANALYSIS
# ORIGINAL TEST vs CLEAN TEST
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load existing object-level CSV files
# ------------------------------------------------------------
original_path = Path(
    "/content/BLPD/Original_Test/object_analysis/"
    "original_test_object_analysis.csv"
)

clean_path = Path(
    "/content/BLPD/Clean_Test/object_analysis/"
    "clean_test_object_analysis.csv"
)

original_df = pd.read_csv(original_path)
clean_df = pd.read_csv(clean_path)

print("Original-test objects:", len(original_df))
print("Clean-test objects:", len(clean_df))


# ------------------------------------------------------------
# 2. Define size groups
# ------------------------------------------------------------
original_df["size_category"] = np.where(
    original_df["min_dimension_px"] < 32,
    "<32 px",
    "≥32 px"
)

clean_df["size_category"] = np.where(
    clean_df["min_dimension_px"] < 32,
    "<32 px",
    "≥32 px"
)


# ------------------------------------------------------------
# 3. Character names
# ------------------------------------------------------------
# Load class names from the dataset YAML
import yaml

yaml_path = "/content/BLPD/LPNR_data.yaml"

with open(yaml_path, "r") as f:
    data_yaml = yaml.safe_load(f)

names = data_yaml["names"]

if isinstance(names, dict):
    class_names = {
        int(k): str(v) for k, v in names.items()
    }
else:
    class_names = {
        i: str(name) for i, name in enumerate(names)
    }


# ------------------------------------------------------------
# 4. Function for per-character analysis
# ------------------------------------------------------------
def analyze_character_vulnerability(df, dataset_name):

    rows = []

    for class_id in sorted(df["true_class"].unique()):

        class_df = df[df["true_class"] == class_id]

        small = class_df[
            class_df["size_category"] == "<32 px"
        ]

        normal = class_df[
            class_df["size_category"] == "≥32 px"
        ]

        small_count = len(small)
        normal_count = len(normal)

        small_acc = (
            small["correct_class"].mean() * 100
            if small_count > 0 else np.nan
        )

        normal_acc = (
            normal["correct_class"].mean() * 100
            if normal_count > 0 else np.nan
        )

        accuracy_drop = (
            normal_acc - small_acc
            if small_count > 0 and normal_count > 0
            else np.nan
        )

        rows.append({

            "dataset": dataset_name,

            "class_id": int(class_id),
            "character": class_names.get(
                int(class_id), str(class_id)
            ),

            "objects_small": small_count,
            "objects_normal": normal_count,

            "accuracy_small": small_acc,
            "accuracy_normal": normal_acc,

            "accuracy_drop": accuracy_drop,

            "small_errors": (
                int((small["correct_class"] == 0).sum())
            ),

            "normal_errors": (
                int((normal["correct_class"] == 0).sum())
            )
        })

    return pd.DataFrame(rows)


# ------------------------------------------------------------
# 5. Analyze original and clean test
# ------------------------------------------------------------
original_results = analyze_character_vulnerability(
    original_df,
    "Original test"
)

clean_results = analyze_character_vulnerability(
    clean_df,
    "Clean test"
)


# ------------------------------------------------------------
# 6. Display the most vulnerable characters
# ------------------------------------------------------------
def display_vulnerable_classes(results, dataset_name):

    print("\n" + "=" * 100)
    print(dataset_name.upper())
    print("CHARACTERS WITH THE LARGEST ACCURACY DROP")
    print("=" * 100)

    # Require at least 5 small and 20 normal examples
    filtered = results[
        (results["objects_small"] >= 5) &
        (results["objects_normal"] >= 20)
    ].copy()

    filtered = filtered.sort_values(
        "accuracy_drop",
        ascending=False
    )

    display(
        filtered.head(30).style.format({
            "accuracy_small": "{:.2f}%",
            "accuracy_normal": "{:.2f}%",
            "accuracy_drop": "{:+.2f}"
        })
    )

    print("\nTotal qualifying character classes:", len(filtered))


display_vulnerable_classes(
    original_results,
    "Original test"
)

display_vulnerable_classes(
    clean_results,
    "Clean test"
)


# ------------------------------------------------------------
# 7. Combine both datasets
# ------------------------------------------------------------
combined = original_results.merge(
    clean_results,
    on=["class_id", "character"],
    suffixes=("_original", "_clean")
)

combined["consistent_accuracy_drop"] = (
    combined["accuracy_drop_original"] +
    combined["accuracy_drop_clean"]
) / 2

# Require sufficient observations in both datasets
combined_filtered = combined[
    (combined["objects_small_original"] >= 5) &
    (combined["objects_small_clean"] >= 5) &
    (combined["objects_normal_original"] >= 20) &
    (combined["objects_normal_clean"] >= 20)
].copy()

combined_filtered = combined_filtered.sort_values(
    "consistent_accuracy_drop",
    ascending=False
)

print("\n" + "=" * 110)
print("CONSISTENTLY VULNERABLE CHARACTERS")
print("=" * 110)

display(
    combined_filtered[[
        "class_id",
        "character",

        "objects_small_original",
        "accuracy_small_original",
        "accuracy_normal_original",
        "accuracy_drop_original",

        "objects_small_clean",
        "accuracy_small_clean",
        "accuracy_normal_clean",
        "accuracy_drop_clean",

        "consistent_accuracy_drop"
    ]].head(30).style.format({
        "accuracy_small_original": "{:.2f}%",
        "accuracy_normal_original": "{:.2f}%",
        "accuracy_drop_original": "{:+.2f}",

        "accuracy_small_clean": "{:.2f}%",
        "accuracy_normal_clean": "{:.2f}%",
        "accuracy_drop_clean": "{:+.2f}",

        "consistent_accuracy_drop": "{:+.2f}"
    })
)


# ------------------------------------------------------------
# 8. Save all results
# ------------------------------------------------------------
output_dir = Path("/content/BLPD/Character_Size_Analysis")
output_dir.mkdir(parents=True, exist_ok=True)

original_results.to_csv(
    output_dir / "original_character_size_analysis.csv",
    index=False
)

clean_results.to_csv(
    output_dir / "clean_character_size_analysis.csv",
    index=False
)

combined.to_csv(
    output_dir / "combined_character_size_analysis.csv",
    index=False
)

print("\nAll analysis files saved to:")
print(output_dir)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# --------------------------------------------------
# 1. Load existing object-analysis CSV files
# --------------------------------------------------

original_path = Path(
    "/content/BLPD/Original_Test/object_analysis/"
    "original_test_object_analysis.csv"
)

clean_path = Path(
    "/content/BLPD/Clean_Test/object_analysis/"
    "clean_test_object_analysis.csv"
)

original_df = pd.read_csv(original_path)
clean_df = pd.read_csv(clean_path)

DIGITS = [str(i) for i in range(10)]

# --------------------------------------------------
# 2. Define width and height groups
# --------------------------------------------------

# These ranges allow us to inspect the smallest
# character dimensions more closely.

width_bins = [0, 8, 12, 16, 24, 32, float("inf")]
width_labels = [
    "<8", "8–11", "12–15", "16–23", "24–31", "≥32"
]

height_bins = [0, 8, 12, 16, 24, 32, float("inf")]
height_labels = [
    "<8", "8–11", "12–15", "16–23", "24–31", "≥32"
]

# --------------------------------------------------
# 3. Prepare the data
# --------------------------------------------------

def prepare_data(df, dataset_name):

    df = df.copy()

    df["true_class"] = df["true_class"].astype(str)
    df["pred_class"] = df["pred_class"].astype(str)

    # Keep numeric characters only
    df = df[df["true_class"].isin(DIGITS)].copy()

    # Add width and height categories
    df["width_group"] = pd.cut(
        df["width_px"],
        bins=width_bins,
        labels=width_labels,
        right=False
    )

    df["height_group"] = pd.cut(
        df["height_px"],
        bins=height_bins,
        labels=height_labels,
        right=False
    )

    df["dataset"] = dataset_name

    return df


original = prepare_data(original_df, "Original Test")
clean = prepare_data(clean_df, "Clean Test")

# --------------------------------------------------
# 4. Calculate accuracy by width or height group
# --------------------------------------------------

def dimension_analysis(df, dimension, dataset_name):

    result = (
        df.groupby(dimension, observed=False)
        .agg(
            total_characters=("correct_class", "size"),
            correctly_classified=("correct_class", "sum")
        )
        .reset_index()
    )

    result["accuracy_percent"] = (
        result["correctly_classified"] /
        result["total_characters"] * 100
    )

    result["error_percent"] = (
        100 - result["accuracy_percent"]
    )

    result.insert(0, "dataset", dataset_name)

    return result.round(2)


# Width analysis
original_width = dimension_analysis(
    original, "width_group", "Original Test"
)

clean_width = dimension_analysis(
    clean, "width_group", "Clean Test"
)

# Height analysis
original_height = dimension_analysis(
    original, "height_group", "Original Test"
)

clean_height = dimension_analysis(
    clean, "height_group", "Clean Test"
)

# --------------------------------------------------
# 5. Display results
# --------------------------------------------------

print("\n========== WIDTH ANALYSIS ==========")

print("\nOriginal Test:")
display(original_width)

print("\nClean Test:")
display(clean_width)


print("\n========== HEIGHT ANALYSIS ==========")

print("\nOriginal Test:")
display(original_height)

print("\nClean Test:")
display(clean_height)

# --------------------------------------------------
# 6. Analyze width and height jointly
# --------------------------------------------------

def joint_analysis(df, dataset_name):

    result = (
        df.groupby(
            ["width_group", "height_group"],
            observed=False
        )
        .agg(
            total_characters=("correct_class", "size"),
            correctly_classified=("correct_class", "sum")
        )
        .reset_index()
    )

    result["accuracy_percent"] = (
        result["correctly_classified"] /
        result["total_characters"] * 100
    )

    result["error_percent"] = (
        100 - result["accuracy_percent"]
    )

    result.insert(0, "dataset", dataset_name)

    # Exclude combinations without observations
    result = result[result["total_characters"] > 0]

    return result.round(2)


original_joint = joint_analysis(original, "Original Test")
clean_joint = joint_analysis(clean, "Clean Test")

print("\n========== JOINT WIDTH × HEIGHT ANALYSIS ==========")

print("\nOriginal Test:")
display(
    original_joint.sort_values(
        "total_characters", ascending=False
    )
)

print("\nClean Test:")
display(
    clean_joint.sort_values(
        "total_characters", ascending=False
    )
)

# --------------------------------------------------
# 7. Save all results
# --------------------------------------------------

output_dir = Path("/content/BLPD/Dimension_Analysis")
output_dir.mkdir(parents=True, exist_ok=True)

original_width.to_csv(
    output_dir / "original_width_analysis.csv", index=False
)

clean_width.to_csv(
    output_dir / "clean_width_analysis.csv", index=False
)

original_height.to_csv(
    output_dir / "original_height_analysis.csv", index=False
)

clean_height.to_csv(
    output_dir / "clean_height_analysis.csv", index=False
)

original_joint.to_csv(
    output_dir / "original_width_height_analysis.csv", index=False
)

clean_joint.to_csv(
    output_dir / "clean_width_height_analysis.csv", index=False
)

print(f"\nAll results saved to: {output_dir}")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

# --------------------------------------------------
# 1. Load CSV files
# --------------------------------------------------

original_path = Path(
    "/content/BLPD/Original_Test/object_analysis/"
    "original_test_object_analysis.csv"
)

clean_path = Path(
    "/content/BLPD/Clean_Test/object_analysis/"
    "clean_test_object_analysis.csv"
)

original_df = pd.read_csv(original_path)
clean_df = pd.read_csv(clean_path)

# --------------------------------------------------
# 2. Failure analysis function
# --------------------------------------------------

def analyze_failures(df, dataset_name):

    df = df.copy()

    # Convert correct_class safely to Boolean
    df["correct_class"] = (
        pd.to_numeric(df["correct_class"], errors="coerce")
        .fillna(0)
        .eq(1)
    )

    # Identify missing predictions
    df["pred_class"] = df["pred_class"].astype(str).str.strip()

    df["missed_detection"] = (
        df["pred_class"].isin(["-1", "nan", "None", ""])
    )

    # Size groups
    bins = [0, 16, 32, 48, 64, 96, float("inf")]
    labels = ["<16", "16–31", "32–47", "48–63", "64–95", "≥96"]

    df["size_group"] = pd.cut(
        df["min_dimension_px"],
        bins=bins,
        labels=labels,
        right=False
    )

    # Classification errors
    df["classification_error"] = (
        (~df["missed_detection"]) &
        (~df["correct_class"])
    )

    # Localization errors:
    # Correct class, but IoU below 0.50
    df["localization_error"] = (
        (~df["missed_detection"]) &
        df["correct_class"] &
        (df["best_iou"] < 0.50)
    )

    # Successful detections
    df["successful_detection"] = (
        (~df["missed_detection"]) &
        df["correct_class"] &
        (df["best_iou"] >= 0.50)
    )

    # --------------------------------------------------
    # Aggregate results
    # --------------------------------------------------

    result = (
        df.groupby("size_group", observed=False)
        .agg(
            total_objects=("gt_index", "size"),
            missed_detections=("missed_detection", "sum"),
            classification_errors=("classification_error", "sum"),
            localization_errors=("localization_error", "sum"),
            successful_detections=("successful_detection", "sum")
        )
        .reset_index()
    )

    categories = [
        "missed_detections",
        "classification_errors",
        "localization_errors",
        "successful_detections"
    ]

    for col in categories:
        result[col + "_percent"] = (
            result[col] / result["total_objects"] * 100
        )

    result.insert(0, "dataset", dataset_name)

    return df, result.round(2)


# --------------------------------------------------
# 3. Run analysis
# --------------------------------------------------

original_processed, original_results = analyze_failures(
    original_df, "Original Test"
)

clean_processed, clean_results = analyze_failures(
    clean_df, "Clean Test"
)

# --------------------------------------------------
# 4. Display complete results
# --------------------------------------------------

print("\n========== ORIGINAL TEST ==========")
display(original_results)

print("\n========== CLEAN TEST ==========")
display(clean_results)

# --------------------------------------------------
# 5. Focus on small characters
# --------------------------------------------------

small_groups = ["<16", "16–31"]

print("\n========== SMALL CHARACTERS: ORIGINAL ==========")
display(
    original_results[
        original_results["size_group"].astype(str).isin(small_groups)
    ]
)

print("\n========== SMALL CHARACTERS: CLEAN ==========")
display(
    clean_results[
        clean_results["size_group"].astype(str).isin(small_groups)
    ]
)

# --------------------------------------------------
# 6. Save results
# --------------------------------------------------

output_dir = Path("/content/BLPD/Failure_Type_Analysis")
output_dir.mkdir(parents=True, exist_ok=True)

original_results.to_csv(
    output_dir / "original_failure_type_analysis.csv",
    index=False
)

clean_results.to_csv(
    output_dir / "clean_failure_type_analysis.csv",
    index=False
)

original_processed.to_csv(
    output_dir / "original_failure_type_object_level.csv",
    index=False
)

clean_processed.to_csv(
    output_dir / "clean_failure_type_object_level.csv",
    index=False
)

print(f"\nResults saved to: {output_dir}")
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image

base = "/content/BLPD"
image_dir = f"{base}/LPNR_Dataset/test/images"
label_dir = f"{base}/LPNR_Dataset/test/labels"

gt_csv = f"{base}/Original_Test/object_analysis/original_test_object_analysis.csv"
pred_csv = f"{base}/small_character_raw_predictions.csv"

gt_df = pd.read_csv(gt_csv)
pred_df = pd.read_csv(pred_csv)

# Only the 84 small ground-truth characters
small_df = gt_df[gt_df["min_dimension_px"] < 16].copy()

def calculate_iou(box, boxes):
    """IoU between one xyxy box and an array of xyxy boxes."""
    if len(boxes) == 0:
        return np.array([])

    x1 = np.maximum(box[0], boxes[:, 0])
    y1 = np.maximum(box[1], boxes[:, 1])
    x2 = np.minimum(box[2], boxes[:, 2])
    y2 = np.minimum(box[3], boxes[:, 3])

    intersection = np.maximum(0, x2-x1) * np.maximum(0, y2-y1)

    area1 = max(0, box[2]-box[0]) * max(0, box[3]-box[1])
    area2 = np.maximum(0, boxes[:, 2]-boxes[:, 0]) * \
            np.maximum(0, boxes[:, 3]-boxes[:, 1])

    union = area1 + area2 - intersection
    return np.divide(
        intersection, union,
        out=np.zeros_like(intersection, dtype=float),
        where=union > 0
    )

rows = []

for _, gt in small_df.iterrows():
    image_name = gt["image"]
    image_path = os.path.join(image_dir, image_name)
    label_path = os.path.join(label_dir, os.path.splitext(image_name)[0] + ".txt")

    # Ground-truth box from the corresponding YOLO label line
    with Image.open(image_path) as im:
        img_w, img_h = im.size

    labels = []
    with open(label_path, "r") as f:
        for line in f:
            values = line.strip().split()
            if len(values) >= 5:
                labels.append([float(x) for x in values[:5]])

    gt_idx = int(gt["gt_index"])
    cls, cx, cy, w, h = labels[gt_idx]

    gt_box = [
        (cx - w/2) * img_w,
        (cy - h/2) * img_h,
        (cx + w/2) * img_w,
        (cy + h/2) * img_h
    ]

    # Predictions from this image
    p = pred_df[pred_df["image"] == image_name]

    if len(p):
        pred_boxes = p[["x1", "y1", "x2", "y2"]].to_numpy()
        pred_classes = p["pred_class"].to_numpy()
        pred_conf = p["confidence"].to_numpy()

        ious = calculate_iou(gt_box, pred_boxes)

        best_idx = int(np.argmax(ious))
        best_iou = float(ious[best_idx])
        best_class = int(pred_classes[best_idx])
        best_conf = float(pred_conf[best_idx])

        same_class = pred_classes == int(cls)
        best_same_iou = float(ious[same_class].max()) if same_class.any() else 0.0

        count_01 = int((ious >= 0.1).sum())
        count_03 = int((ious >= 0.3).sum())
        count_05 = int((ious >= 0.5).sum())
    else:
        best_iou = 0.0
        best_class = -1
        best_conf = 0.0
        best_same_iou = 0.0
        count_01 = count_03 = count_05 = 0

    rows.append({
        "image": image_name,
        "gt_index": gt_idx,
        "true_class": int(cls),
        "width_px": gt["width_px"],
        "height_px": gt["height_px"],
        "previous_pred_class": gt["pred_class"],
        "previous_best_iou": gt["best_iou"],
        "best_raw_iou": best_iou,
        "best_raw_pred_class": best_class,
        "best_raw_confidence": best_conf,
        "best_correct_class_iou": best_same_iou,
        "predictions_iou_01": count_01,
        "predictions_iou_03": count_03,
        "predictions_iou_05": count_05
    })

inspection = pd.DataFrame(rows)

out_path = f"{base}/tiny_character_prediction_inspection.csv"
inspection.to_csv(out_path, index=False)

# Summary
print("GROUND-TRUTH CHARACTER INSPECTION")
print("Total:", len(inspection))

print("\nBest prediction IoU:")
print(inspection["best_raw_iou"].describe())

print("\nBest correct-class IoU:")
print(inspection["best_correct_class_iou"].describe())

print("\nOverlap counts:")
for threshold in [0.1, 0.3, 0.5]:
    col = f"predictions_iou_{int(threshold*10):02d}"
    print(f"GT characters with at least one prediction at IoU >= {threshold}:",
          (inspection[col] > 0).sum())

# Focus on the 70 characters previously marked as missed or incorrect
print("\nPREVIOUSLY MISSED CHARACTERS")
previously_missed = inspection[
    inspection["previous_pred_class"] == -1
]

print("Count:", len(previously_missed))
print(previously_missed[[
    "image", "gt_index", "true_class",
    "best_raw_iou", "best_raw_pred_class",
    "best_raw_confidence", "best_correct_class_iou",
    "predictions_iou_01"
]].to_string(index=False))

print("\nSaved:", out_path)
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image

base = "/content/BLPD"
image_dir = f"{base}/LPNR_Dataset/test/images"
label_dir = f"{base}/LPNR_Dataset/test/labels"

inspection = pd.read_csv(
    f"{base}/tiny_character_prediction_inspection.csv"
)

def calculate_iou(box, boxes):
    if len(boxes) == 0:
        return np.array([])

    x1 = np.maximum(box[0], boxes[:, 0])
    y1 = np.maximum(box[1], boxes[:, 1])
    x2 = np.minimum(box[2], boxes[:, 2])
    y2 = np.minimum(box[3], boxes[:, 3])

    intersection = np.maximum(0, x2-x1) * np.maximum(0, y2-y1)

    area1 = max(0, box[2]-box[0]) * max(0, box[3]-box[1])
    area2 = np.maximum(0, boxes[:, 2]-boxes[:, 0]) * \
            np.maximum(0, boxes[:, 3]-boxes[:, 1])

    union = area1 + area2 - intersection

    return np.divide(
        intersection, union,
        out=np.zeros_like(intersection, dtype=float),
        where=union > 0
    )

all_summaries = []
all_details = []

for imgsz in [640, 960, 1280]:

    pred_df = pd.read_csv(
        f"{base}/tiny_predictions_{imgsz}.csv"
    )

    rows = []

    for _, gt in inspection.iterrows():

        image_name = gt["image"]
        image_path = os.path.join(image_dir, image_name)
        label_path = os.path.join(
            label_dir, os.path.splitext(image_name)[0] + ".txt"
        )

        with Image.open(image_path) as im:
            img_w, img_h = im.size

        with open(label_path) as f:
            labels = [
                [float(v) for v in line.strip().split()[:5]]
                for line in f if len(line.strip().split()) >= 5
            ]

        cls, cx, cy, w, h = labels[int(gt["gt_index"])]

        gt_box = np.array([
            (cx-w/2)*img_w,
            (cy-h/2)*img_h,
            (cx+w/2)*img_w,
            (cy+h/2)*img_h
        ])

        p = pred_df[pred_df["image"] == image_name]

        if len(p):
            boxes = p[["x1", "y1", "x2", "y2"]].to_numpy()
            classes = p["pred_class"].to_numpy().astype(int)
            confs = p["confidence"].to_numpy()

            ious = calculate_iou(gt_box, boxes)

            best_idx = int(np.argmax(ious))
            best_iou = float(ious[best_idx])
            best_class = int(classes[best_idx])
            best_conf = float(confs[best_idx])

            correct_mask = classes == int(cls)

            if correct_mask.any():
                correct_indices = np.where(correct_mask)[0]
                correct_idx = correct_indices[
                    np.argmax(ious[correct_mask])
                ]
                correct_iou = float(ious[correct_idx])
                correct_conf = float(confs[correct_idx])
            else:
                correct_iou = 0.0
                correct_conf = 0.0

        else:
            best_iou = 0.0
            best_class = -1
            best_conf = 0.0
            correct_iou = 0.0
            correct_conf = 0.0

        rows.append({
            "imgsz": imgsz,
            "image": image_name,
            "gt_index": int(gt["gt_index"]),
            "true_class": int(cls),
            "best_iou": best_iou,
            "best_pred_class": best_class,
            "best_confidence": best_conf,
            "best_correct_class_iou": correct_iou,
            "best_correct_class_confidence": correct_conf,
            "best_class_correct": best_class == int(cls)
        })

    result_df = pd.DataFrame(rows)
    all_details.append(result_df)

    all_summaries.append({
        "resolution": imgsz,
        "mean_best_iou": result_df["best_iou"].mean(),
        "mean_correct_class_iou": result_df[
            "best_correct_class_iou"
        ].mean(),
        "best_prediction_class_correct_pct":
            100 * result_df["best_class_correct"].mean(),
        "correct_class_iou_50_pct":
            100 * (result_df["best_correct_class_iou"] >= 0.5).mean(),
        "correct_class_iou_75_pct":
            100 * (result_df["best_correct_class_iou"] >= 0.75).mean(),
        "characters_with_correct_class_prediction":
            (result_df["best_correct_class_iou"] > 0).sum()
    })

summary = pd.DataFrame(all_summaries)
details = pd.concat(all_details, ignore_index=True)

summary.to_csv(
    f"{base}/resolution_comparison_summary.csv", index=False
)
details.to_csv(
    f"{base}/resolution_comparison_details.csv", index=False
)

print("RESOLUTION COMPARISON")
print(summary.round(4).to_string(index=False))

print("\nSaved summary and detailed results to:", base)
```

```python
import pandas as pd
import numpy as np
import cv2
from pathlib import Path
from collections import defaultdict

# ============================================================
# Paths
# ============================================================

base_dir = Path("/content/BLPD/Clean_Test")
image_dir = base_dir / "images"
label_dir = base_dir / "labels"

pred_640_path = (
    base_dir / "resolution_640" / "predictions_640.csv"
)

pred_1280_path = (
    base_dir / "resolution_1280" / "predictions_1280.csv"
)

output_dir = base_dir / "paired_640_vs_1280"
output_dir.mkdir(parents=True, exist_ok=True)


# ============================================================
# Load predictions
# ============================================================

pred640_df = pd.read_csv(pred_640_path)
pred1280_df = pd.read_csv(pred_1280_path)

print("640 px predictions:", len(pred640_df))
print("1280 px predictions:", len(pred1280_df))


# ============================================================
# Group predictions by image
# ============================================================

def group_predictions(df):
    predictions = defaultdict(list)

    for _, row in df.iterrows():
        predictions[row["image"]].append({
            "box": np.array([
                row["x1"],
                row["y1"],
                row["x2"],
                row["y2"]
            ], dtype=float),
            "class": int(row["pred_class"]),
            "confidence": float(row["confidence"])
        })

    return predictions


predictions_640 = group_predictions(pred640_df)
predictions_1280 = group_predictions(pred1280_df)


# ============================================================
# IoU function
# ============================================================

def calculate_iou(box1, box2):

    x_left = max(box1[0], box2[0])
    y_top = max(box1[1], box2[1])
    x_right = min(box1[2], box2[2])
    y_bottom = min(box1[3], box2[3])

    intersection = (
        max(0, x_right - x_left) *
        max(0, y_bottom - y_top)
    )

    area1 = (
        max(0, box1[2] - box1[0]) *
        max(0, box1[3] - box1[1])
    )

    area2 = (
        max(0, box2[2] - box2[0]) *
        max(0, box2[3] - box2[1])
    )

    union = area1 + area2 - intersection

    return intersection / union if union > 0 else 0.0


# ============================================================
# Size groups
# ============================================================

def get_size_group(min_dimension):

    if min_dimension < 16:
        return "<16"
    elif min_dimension < 32:
        return "16–31"
    elif min_dimension < 48:
        return "32–47"
    elif min_dimension < 64:
        return "48–63"
    elif min_dimension < 96:
        return "64–95"
    else:
        return "≥96"


# ============================================================
# Find best prediction for one ground-truth box
# ============================================================

def find_best_prediction(gt_box, image_predictions):

    best_iou = 0.0
    best_class = -1
    best_confidence = 0.0

    for pred in image_predictions:

        iou = calculate_iou(gt_box, pred["box"])

        if iou > best_iou:

            best_iou = iou
            best_class = pred["class"]
            best_confidence = pred["confidence"]

    return best_iou, best_class, best_confidence


# ============================================================
# Paired analysis
# ============================================================

paired_rows = []

processed_images = 0
missing_labels = 0

for image_path in sorted(image_dir.iterdir()):

    if image_path.suffix.lower() not in [
        ".jpg", ".jpeg", ".png"
    ]:
        continue

    label_path = label_dir / (image_path.stem + ".txt")

    if not label_path.exists():
        missing_labels += 1
        continue

    image = cv2.imread(str(image_path))

    if image is None:
        continue

    img_height, img_width = image.shape[:2]

    image_name = image_path.name

    preds640 = predictions_640.get(image_name, [])
    preds1280 = predictions_1280.get(image_name, [])

    # --------------------------------------------------------
    # Read ground-truth labels
    # --------------------------------------------------------

    with open(label_path, "r") as f:
        lines = [
            line.strip()
            for line in f
            if line.strip()
        ]

    for gt_index, line in enumerate(lines):

        values = line.split()

        if len(values) != 5:
            continue

        gt_class = int(float(values[0]))

        x_center, y_center, width, height = map(
            float,
            values[1:]
        )

        # YOLO normalized coordinates -> original image pixels
        gt_box = np.array([
            (x_center - width / 2) * img_width,
            (y_center - height / 2) * img_height,
            (x_center + width / 2) * img_width,
            (y_center + height / 2) * img_height
        ])

        width_px = gt_box[2] - gt_box[0]
        height_px = gt_box[3] - gt_box[1]

        min_dimension = min(
            width_px,
            height_px
        )

        size_group = get_size_group(
            min_dimension
        )

        # ----------------------------------------------------
        # 640 px best prediction
        # ----------------------------------------------------

        iou640, class640, conf640 = find_best_prediction(
            gt_box,
            preds640
        )

        # ----------------------------------------------------
        # 1280 px best prediction
        # ----------------------------------------------------

        iou1280, class1280, conf1280 = find_best_prediction(
            gt_box,
            preds1280
        )

        # ----------------------------------------------------
        # Correctness
        # ----------------------------------------------------

        correct_class_640 = int(
            class640 == gt_class
        )

        correct_class_1280 = int(
            class1280 == gt_class
        )

        iou50_640 = int(
            iou640 >= 0.50
        )

        iou50_1280 = int(
            iou1280 >= 0.50
        )

        correct_iou50_640 = int(
            correct_class_640 and iou50_640
        )

        correct_iou50_1280 = int(
            correct_class_1280 and iou50_1280
        )

        # ----------------------------------------------------
        # Paired outcome
        # ----------------------------------------------------

        if (
            correct_iou50_640 == 1
            and correct_iou50_1280 == 1
        ):
            outcome = "correct_both"

        elif (
            correct_iou50_640 == 0
            and correct_iou50_1280 == 1
        ):
            outcome = "improved_1280"

        elif (
            correct_iou50_640 == 1
            and correct_iou50_1280 == 0
        ):
            outcome = "worsened_1280"

        else:
            outcome = "incorrect_both"

        paired_rows.append({

            "image": image_name,
            "gt_index": gt_index,

            "true_class": gt_class,

            "width_px": width_px,
            "height_px": height_px,
            "min_dimension_px": min_dimension,
            "size_group": size_group,

            # 640 px
            "pred_class_640": class640,
            "confidence_640": conf640,
            "best_iou_640": iou640,
            "correct_class_640": correct_class_640,
            "iou50_640": iou50_640,
            "correct_iou50_640": correct_iou50_640,

            # 1280 px
            "pred_class_1280": class1280,
            "confidence_1280": conf1280,
            "best_iou_1280": iou1280,
            "correct_class_1280": correct_class_1280,
            "iou50_1280": iou50_1280,
            "correct_iou50_1280": correct_iou50_1280,

            # Differences
            "iou_change_1280_minus_640":
                iou1280 - iou640,

            "confidence_change_1280_minus_640":
                conf1280 - conf640,

            "outcome": outcome
        })

    processed_images += 1


# ============================================================
# Create DataFrame
# ============================================================

paired_df = pd.DataFrame(paired_rows)

paired_path = (
    output_dir /
    "paired_character_analysis_640_vs_1280.csv"
)

paired_df.to_csv(
    paired_path,
    index=False
)


# ============================================================
# Overall outcome summary
# ============================================================

outcome_summary = (
    paired_df["outcome"]
    .value_counts()
    .rename_axis("outcome")
    .reset_index(name="characters")
)

outcome_summary["percentage"] = (
    outcome_summary["characters"] /
    len(paired_df) * 100
)


# ============================================================
# Size-group outcome summary
# ============================================================

size_summary = (
    paired_df
    .groupby(["size_group", "outcome"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

# Make sure all outcome columns exist
for col in [
    "correct_both",
    "improved_1280",
    "worsened_1280",
    "incorrect_both"
]:
    if col not in size_summary.columns:
        size_summary[col] = 0


# Calculate total characters
size_summary["total"] = (
    size_summary["correct_both"] +
    size_summary["improved_1280"] +
    size_summary["worsened_1280"] +
    size_summary["incorrect_both"]
)

# Calculate percentages
for col in [
    "correct_both",
    "improved_1280",
    "worsened_1280",
    "incorrect_both"
]:
    size_summary[col + "_pct"] = (
        size_summary[col] /
        size_summary["total"] * 100
    )

# Net change
size_summary["net_change_1280"] = (
    size_summary["improved_1280"] -
    size_summary["worsened_1280"]
)


size_summary_path = (
    output_dir /
    "paired_size_group_summary_640_vs_1280.csv"
)

size_summary.to_csv(
    size_summary_path,
    index=False
)


# ============================================================
# Direct performance comparison
# ============================================================

overall_comparison = pd.DataFrame({

    "metric": [
        "Correct class",
        "IoU ≥ 0.50",
        "Correct class + IoU ≥ 0.50",
        "Mean IoU",
        "Mean confidence"
    ],

    "640_px": [
        paired_df["correct_class_640"].mean() * 100,
        paired_df["iou50_640"].mean() * 100,
        paired_df["correct_iou50_640"].mean() * 100,
        paired_df["best_iou_640"].mean(),
        paired_df["confidence_640"].mean()
    ],

    "1280_px": [
        paired_df["correct_class_1280"].mean() * 100,
        paired_df["iou50_1280"].mean() * 100,
        paired_df["correct_iou50_1280"].mean() * 100,
        paired_df["best_iou_1280"].mean(),
        paired_df["confidence_1280"].mean()
    ]
})

overall_comparison["change_1280_minus_640"] = (
    overall_comparison["1280_px"] -
    overall_comparison["640_px"]
)

comparison_path = (
    output_dir /
    "overall_comparison_640_vs_1280.csv"
)

overall_comparison.to_csv(
    comparison_path,
    index=False
)


# ============================================================
# Print results
# ============================================================

print("=" * 70)
print("PAIRED 640 vs 1280 ANALYSIS")
print("=" * 70)

print("\nImages processed:", processed_images)
print("Missing labels:", missing_labels)
print("Ground-truth characters:", len(paired_df))

print("\nOutcome summary:")
display(outcome_summary.round(3))

print("\nSize-group outcome summary:")
display(size_summary.round(3))

print("\nOverall comparison:")
display(overall_comparison.round(4))

print("\nSaved files:")
print(paired_path)
print(size_summary_path)
print(comparison_path)
```

```python
# ============================================================
# 2. Separate the 592 worsened cases
# ============================================================

# A worsened case starts as correct_iou50 at 640
# and becomes incorrect_iou50 at 1280.

w = df[df["outcome"] == "worsened_1280"].copy()

# ------------------------------------------------------------
# Define failure types at 1280
# ------------------------------------------------------------

# A: Localization is still good, but class becomes wrong
w["classification_only_failure"] = (
    (w["iou50_1280"] == 1) &
    (w["correct_class_1280"] == 0)
)

# B: Localization drops below IoU 0.50, but predicted class
# is still the correct class
w["localization_only_failure"] = (
    (w["iou50_1280"] == 0) &
    (w["correct_class_1280"] == 1)
)

# C: Both class and localization become wrong
w["both_failure"] = (
    (w["iou50_1280"] == 0) &
    (w["correct_class_1280"] == 0)
)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

failure_counts = pd.Series({
    "Classification only": w["classification_only_failure"].sum(),
    "Localization only": w["localization_only_failure"].sum(),
    "Both classification + localization": w["both_failure"].sum()
})

failure_summary = pd.DataFrame({
    "Count": failure_counts,
    "Percentage": failure_counts / len(w) * 100
})

print(failure_summary.round(2))
print("\nTotal:", failure_counts.sum())
```

```python
# ============================================================
# 3. Class-confusion analysis
# ============================================================

# Keep only cases where localization at 1280 is still valid
# (IoU >= 0.50), so that we are studying classification
# rather than localization.

class_confusions = w[
    w["classification_only_failure"]
].copy()

print("Classification-only failures:", len(class_confusions))

# ------------------------------------------------------------
# True class -> 1280 predicted class
# ------------------------------------------------------------

confusion_table = (
    class_confusions
    .groupby(["true_class", "pred_class_1280"])
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
)

print("\nTop class confusions:")
display(confusion_table.head(30))
```

```python
# ============================================================
# 4. 640 correct -> 1280 wrong transition analysis
# ============================================================

transition_table = (
    class_confusions
    .groupby(["true_class", "pred_class_640", "pred_class_1280"])
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
)

print("Top 640 -> 1280 class transitions:")
display(transition_table.head(40))
```

```python
# ============================================================
# 5. Failure type by character size
# ============================================================

size_failure = (
    w.groupby("size_group")
    .agg(
        total=("true_class", "size"),
        classification_only=("classification_only_failure", "sum"),
        localization_only=("localization_only_failure", "sum"),
        both=("both_failure", "sum")
    )
    .reset_index()
)

for col in ["classification_only", "localization_only", "both"]:
    size_failure[col + "_pct"] = (
        size_failure[col] / size_failure["total"] * 100
    )

display(size_failure)
```

```python
# ============================================================
# 6. Confidence and IoU changes
# ============================================================

classification_only = w[
    w["classification_only_failure"]
].copy()

print("Classification-only failures:", len(classification_only))

print("\nConfidence:")
print(
    classification_only[
        ["confidence_640", "confidence_1280",
         "confidence_change_1280_minus_640"]
    ].describe().round(4)
)

print("\nIoU:")
print(
    classification_only[
        ["best_iou_640", "best_iou_1280",
         "iou_change_1280_minus_640"]
    ].describe().round(4)
)
```

```python
# Cell 7: Per-class 640 -> 1280 degradation analysis

import pandas as pd

# Use the full paired dataset
class_analysis = (
    df.groupby("true_class")
    .agg(
        total=("true_class", "size"),
        correct_640=("correct_class_640", "sum"),
        correct_1280=("correct_class_1280", "sum"),
        correct_iou50_640=("correct_iou50_640", "sum"),
        correct_iou50_1280=("correct_iou50_1280", "sum")
    )
    .reset_index()
)

# Classification accuracy for each class
class_analysis["acc_640_pct"] = (
    class_analysis["correct_640"] /
    class_analysis["total"] * 100
)

class_analysis["acc_1280_pct"] = (
    class_analysis["correct_1280"] /
    class_analysis["total"] * 100
)

# Change in classification accuracy
class_analysis["classification_change_pp"] = (
    class_analysis["acc_1280_pct"] -
    class_analysis["acc_640_pct"]
)

# Number of cases that changed from correct at 640 to incorrect at 1280
class_analysis["worsened_count"] = (
    (df["outcome"] == "worsened_1280")
    .groupby(df["true_class"])
    .sum()
    .reindex(class_analysis["true_class"])
    .values
)

# Worsening rate relative to the number of instances of each class
class_analysis["worsened_pct"] = (
    class_analysis["worsened_count"] /
    class_analysis["total"] * 100
)

# Sort by number of worsened cases
class_analysis = class_analysis.sort_values(
    "worsened_count",
    ascending=False
)

print("Top classes affected by 1280 resolution:")
display(
    class_analysis[
        [
            "true_class",
            "total",
            "correct_640",
            "correct_1280",
            "acc_640_pct",
            "acc_1280_pct",
            "classification_change_pp",
            "worsened_count",
            "worsened_pct"
        ]
    ].head(30).round(2)
)
```

```python
# Cell 8: Class frequency vs. resolution-induced degradation

import numpy as np
import pandas as pd

# Use the per-class results from Cell 7
analysis = class_analysis.copy()

# Pearson and Spearman correlations
pearson_corr = analysis["total"].corr(
    analysis["worsened_pct"],
    method="pearson"
)

spearman_corr = analysis["total"].corr(
    analysis["worsened_pct"],
    method="spearman"
)

print(f"Pearson correlation (class frequency vs. worsening rate): {pearson_corr:.4f}")
print(f"Spearman correlation (class frequency vs. worsening rate): {spearman_corr:.4f}")

# Show the most frequent classes
print("\nMost frequent classes:")
display(
    analysis[
        [
            "true_class",
            "total",
            "worsened_count",
            "worsened_pct",
            "classification_change_pp"
        ]
    ]
    .sort_values("total", ascending=False)
    .head(20)
    .round(2)
)

# Show classes with the highest worsening rate,
# but require at least 20 instances to avoid tiny-sample effects
print("\nHighest worsening rates among classes with >=20 instances:")
display(
    analysis[analysis["total"] >= 20][
        [
            "true_class",
            "total",
            "worsened_count",
            "worsened_pct",
            "classification_change_pp"
        ]
    ]
    .sort_values("worsened_pct", ascending=False)
    .head(20)
    .round(2)
)
```

```python
# Cell 9: Identify classes that benefit from 1280 resolution

improved_analysis = class_analysis.copy()

# Number of cases that were incorrect at 640 but correct at 1280
improved_counts = (
    (df["outcome"] == "improved_1280")
    .groupby(df["true_class"])
    .sum()
)

improved_analysis["improved_count"] = (
    improved_analysis["true_class"]
    .map(improved_counts)
    .fillna(0)
    .astype(int)
)

improved_analysis["improved_pct"] = (
    improved_analysis["improved_count"] /
    improved_analysis["total"] * 100
)

improved_analysis["net_change_count"] = (
    improved_analysis["improved_count"] -
    improved_analysis["worsened_count"]
)

# Sort by net benefit
print("Classes with the greatest net benefit at 1280:")
display(
    improved_analysis[
        [
            "true_class",
            "total",
            "improved_count",
            "worsened_count",
            "net_change_count",
            "acc_640_pct",
            "acc_1280_pct",
            "classification_change_pp"
        ]
    ]
    .sort_values("net_change_count", ascending=False)
    .head(25)
    .round(2)
)

# Classes with the greatest number of improvements
print("\nClasses with the most improved predictions:")
display(
    improved_analysis[
        [
            "true_class",
            "total",
            "improved_count",
            "improved_pct",
            "worsened_count",
            "net_change_count"
        ]
    ]
    .sort_values("improved_count", ascending=False)
    .head(20)
    .round(2)
)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from PIL import Image

base = Path("/content/BLPD/Clean_Test")
image_dir = base / "images"
label_dir = base / "labels"

def calculate_iou(box, boxes):
    """Calculate IoU between one box and an array of boxes."""
    x1 = np.maximum(box[0], boxes[:, 0])
    y1 = np.maximum(box[1], boxes[:, 1])
    x2 = np.minimum(box[2], boxes[:, 2])
    y2 = np.minimum(box[3], boxes[:, 3])

    intersection = np.maximum(0, x2 - x1) * np.maximum(0, y2 - y1)

    area_box = (box[2] - box[0]) * (box[3] - box[1])
    area_boxes = (
        (boxes[:, 2] - boxes[:, 0]) *
        (boxes[:, 3] - boxes[:, 1])
    )

    union = area_box + area_boxes - intersection
    return np.divide(
        intersection,
        union,
        out=np.zeros_like(intersection, dtype=float),
        where=union > 0
    )

comparison_rows = []

for _, case in sample.iterrows():

    image_name = case["image"]
    gt_index = int(case["gt_index"])

    # Load original image dimensions
    image_path = image_dir / image_name
    img = Image.open(image_path)
    W, H = img.size

    # Read the corresponding ground-truth annotation
    label_path = label_dir / (Path(image_name).stem + ".txt")
    annotations = label_path.read_text().strip().splitlines()

    parts = annotations[gt_index].split()
    true_class, xc, yc, bw, bh = map(float, parts)

    # Convert YOLO normalized coordinates to original pixel coordinates
    gt_box = np.array([
        (xc - bw / 2) * W,
        (yc - bh / 2) * H,
        (xc + bw / 2) * W,
        (yc + bh / 2) * H
    ])

    # Find predictions for this image at 960 px
    image_preds = pred_960[pred_960["image"] == image_name]

    if image_preds.empty:
        pred_class_960 = np.nan
        confidence_960 = np.nan
        best_iou_960 = 0.0
    else:
        pred_boxes = image_preds[["x1", "y1", "x2", "y2"]].to_numpy()
        ious = calculate_iou(gt_box, pred_boxes)

        best_idx = int(np.argmax(ious))
        best_prediction = image_preds.iloc[best_idx]

        pred_class_960 = int(best_prediction["pred_class"])
        confidence_960 = float(best_prediction["confidence"])
        best_iou_960 = float(ious[best_idx])

    comparison_rows.append({
        "image": image_name,
        "gt_index": gt_index,
        "true_class": int(true_class),
        "size_group": case["size_group"],

        "pred_class_640": case["pred_class_640"],
        "confidence_640": case["confidence_640"],
        "iou_640": case["best_iou_640"],

        "pred_class_960": pred_class_960,
        "confidence_960": confidence_960,
        "iou_960": best_iou_960,

        "pred_class_1280": case["pred_class_1280"],
        "confidence_1280": case["confidence_1280"],
        "iou_1280": case["best_iou_1280"]
    })

comparison_3res = pd.DataFrame(comparison_rows)

# Add classification correctness indicators
for resolution in [640, 960, 1280]:
    comparison_3res[f"correct_{resolution}"] = (
        comparison_3res[f"pred_class_{resolution}"] ==
        comparison_3res["true_class"]
    )

print("Three-resolution comparison shape:", comparison_3res.shape)
print("\nMissing 960 predictions:", comparison_3res["pred_class_960"].isna().sum())

display(comparison_3res[[
    "true_class", "size_group",
    "pred_class_640", "confidence_640", "iou_640",
    "pred_class_960", "confidence_960", "iou_960",
    "pred_class_1280", "confidence_1280", "iou_1280"
]])
```

```python
# Class-specific resolution sensitivity
# Only classes with at least 20 ground-truth instances

class_analysis = (
    full_3res
    .groupby("true_class")
    .agg(
        Total=("true_class", "size"),
        Correct_640=("correct_class_640", "sum"),
        Correct_960=("correct_class_960", "sum"),
        Correct_1280=("correct_class_1280", "sum")
    )
    .reset_index()
)

class_analysis["Accuracy_640"] = (
    class_analysis["Correct_640"] / class_analysis["Total"] * 100
)

class_analysis["Accuracy_960"] = (
    class_analysis["Correct_960"] / class_analysis["Total"] * 100
)

class_analysis["Accuracy_1280"] = (
    class_analysis["Correct_1280"] / class_analysis["Total"] * 100
)

class_analysis["Change_960_vs_640"] = (
    class_analysis["Accuracy_960"] - class_analysis["Accuracy_640"]
)

class_analysis["Change_1280_vs_640"] = (
    class_analysis["Accuracy_1280"] - class_analysis["Accuracy_640"]
)

class_analysis["Improved_1280"] = (
    (full_3res.assign(
        improved=(full_3res["correct_class_640"] == 0) &
                 (full_3res["correct_class_1280"] == 1)
    ))
    .groupby("true_class")["improved"]
    .sum()
    .reindex(class_analysis["true_class"])
    .to_numpy()
)

class_analysis["Worsened_1280"] = (
    (full_3res.assign(
        worsened=(full_3res["correct_class_640"] == 1) &
                 (full_3res["correct_class_1280"] == 0)
    ))
    .groupby("true_class")["worsened"]
    .sum()
    .reindex(class_analysis["true_class"])
    .to_numpy()
)

class_analysis = class_analysis[
    class_analysis["Total"] >= 20
].copy()

class_analysis = class_analysis.round(2)

print("CLASSES WITH THE LARGEST DEGRADATION AT 1280 PX")
display(
    class_analysis.sort_values(
        "Change_1280_vs_640"
    ).head(15)
)

print("\nCLASSES THAT BENEFIT MOST FROM 1280 PX")
display(
    class_analysis.sort_values(
        "Change_1280_vs_640",
        ascending=False
    ).head(15)
)
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image

image_dir = "/content/BLPD/Clean_Test/images"
label_dir = "/content/BLPD/Clean_Test/labels"

# Group 960 px predictions by image
pred_groups_960 = {
    image: group.reset_index(drop=True)
    for image, group in pred_960.groupby("image")
}

# Cache ground-truth labels and original image dimensions
label_cache = {}
image_size_cache = {}

def get_ground_truth(image, gt_index):
    if image not in label_cache:
        label_path = os.path.join(
            label_dir,
            os.path.splitext(image)[0] + ".txt"
        )
        with open(label_path, "r") as f:
            label_cache[image] = [
                line.strip().split()
                for line in f if line.strip()
            ]

    if image not in image_size_cache:
        with Image.open(os.path.join(image_dir, image)) as img:
            image_size_cache[image] = img.size

    label = label_cache[image][int(gt_index)]
    img_width, img_height = image_size_cache[image]

    class_id = int(float(label[0]))
    xc, yc, w, h = map(float, label[1:5])

    gt_box = [
        (xc - w / 2) * img_width,
        (yc - h / 2) * img_height,
        (xc + w / 2) * img_width,
        (yc + h / 2) * img_height
    ]

    return class_id, gt_box


def calculate_iou(box1, box2):
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = max(0, x2 - x1) * max(0, y2 - y1)

    area1 = max(0, box1[2] - box1[0]) * max(0, box1[3] - box1[1])
    area2 = max(0, box2[2] - box2[0]) * max(0, box2[3] - box2[1])

    union = area1 + area2 - intersection

    return intersection / union if union > 0 else 0


# Start from the existing 640 vs 1280 paired analysis
full_3res = df.copy()

pred_classes = []
confidences = []
best_ious = []

for _, row in full_3res.iterrows():

    image = row["image"]
    gt_index = int(row["gt_index"])

    true_class, gt_box = get_ground_truth(image, gt_index)

    predictions = pred_groups_960.get(image)

    if predictions is None or len(predictions) == 0:
        pred_classes.append(-1)
        confidences.append(0.0)
        best_ious.append(0.0)
        continue

    pred_boxes = predictions[
        ["x1", "y1", "x2", "y2"]
    ].to_numpy()

    ious = np.array([
        calculate_iou(gt_box, box)
        for box in pred_boxes
    ])

    best_idx = int(np.argmax(ious))
    best_prediction = predictions.iloc[best_idx]

    pred_classes.append(int(best_prediction["pred_class"]))
    confidences.append(float(best_prediction["confidence"]))
    best_ious.append(float(ious[best_idx]))


# Add 960 px results
full_3res["pred_class_960"] = pred_classes
full_3res["confidence_960"] = confidences
full_3res["best_iou_960"] = best_ious

# Calculate correctness
full_3res["correct_class_640"] = (
    full_3res["pred_class_640"] == full_3res["true_class"]
).astype(int)

full_3res["correct_class_960"] = (
    full_3res["pred_class_960"] == full_3res["true_class"]
).astype(int)

full_3res["correct_class_1280"] = (
    full_3res["pred_class_1280"] == full_3res["true_class"]
).astype(int)

# Calculate localization and combined correctness
for resolution in [640, 960, 1280]:
    full_3res[f"iou50_{resolution}"] = (
        full_3res[f"best_iou_{resolution}"] >= 0.50
    ).astype(int)

    full_3res[f"correct_iou50_{resolution}"] = (
        full_3res[f"correct_class_{resolution}"]
        * full_3res[f"iou50_{resolution}"]
    )

# Display verification
print("Reconstructed full_3res:", full_3res.shape)

print("\nClassification accuracy:")
for resolution in [640, 960, 1280]:
    accuracy = full_3res[f"correct_class_{resolution}"].mean() * 100
    print(f"{resolution}px: {accuracy:.2f}%")

print("\nIoU >= 0.50:")
for resolution in [640, 960, 1280]:
    iou_rate = full_3res[f"iou50_{resolution}"].mean() * 100
    print(f"{resolution}px: {iou_rate:.2f}%")

display(full_3res.head())
```

```python
import cv2
import os
import matplotlib.pyplot as plt
import pandas as pd

# Sort the 24 cases and select the first six for visual inspection
pair = full_3res[
    (full_3res["true_class"] == 5) &
    (full_3res["pred_class_1280"] == 0)
].sort_values(["image", "gt_index"]).reset_index(drop=True)

sample_cases = pair.head(6)

image_dir = "/content/BLPD/Clean_Test/images"
label_dir = "/content/BLPD/Clean_Test/labels"

def letterbox_image(image, target_size):
    h, w = image.shape[:2]

    scale = min(target_size / h, target_size / w)

    new_w = round(w * scale)
    new_h = round(h * scale)

    resized = cv2.resize(
        image,
        (new_w, new_h),
        interpolation=cv2.INTER_LINEAR
    )

    canvas = 114 * __import__("numpy").ones(
        (target_size, target_size, 3),
        dtype="uint8"
    )

    left = (target_size - new_w) // 2
    top = (target_size - new_h) // 2

    canvas[top:top + new_h, left:left + new_w] = resized

    return canvas, scale, left, top


def extract_character_crop(image, gt_box, target_size=None):
    if target_size is None:
        transformed = image
        scale, left, top = 1.0, 0, 0
    else:
        transformed, scale, left, top = letterbox_image(
            image, target_size
        )

    x1, y1, x2, y2 = gt_box

    x1 = int(round(x1 * scale + left))
    y1 = int(round(y1 * scale + top))
    x2 = int(round(x2 * scale + left))
    y2 = int(round(y2 * scale + top))

    # Add 20% context around the ground-truth character
    margin_x = max(2, int((x2 - x1) * 0.20))
    margin_y = max(2, int((y2 - y1) * 0.20))

    x1 = max(0, x1 - margin_x)
    y1 = max(0, y1 - margin_y)
    x2 = min(transformed.shape[1], x2 + margin_x)
    y2 = min(transformed.shape[0], y2 + margin_y)

    crop = transformed[y1:y2, x1:x2]

    return cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)


fig, axes = plt.subplots(6, 4, figsize=(16, 18))

for row_idx, (_, row) in enumerate(sample_cases.iterrows()):

    image_path = os.path.join(image_dir, row["image"])
    image = cv2.imread(image_path)

    _, gt_box = get_ground_truth(
        row["image"],
        row["gt_index"]
    )

    crops = [
        extract_character_crop(image, gt_box),
        extract_character_crop(image, gt_box, 640),
        extract_character_crop(image, gt_box, 960),
        extract_character_crop(image, gt_box, 1280)
    ]

    predictions = [
        int(row["pred_class_640"]),
        int(row["pred_class_640"]),
        int(row["pred_class_960"]),
        int(row["pred_class_1280"])
    ]

    titles = [
        "Original crop",
        f"640 px — predicted {predictions[1]}",
        f"960 px — predicted {predictions[2]}",
        f"1280 px — predicted {predictions[3]}"
    ]

    for col_idx, (crop, title) in enumerate(zip(crops, titles)):
        axes[row_idx, col_idx].imshow(crop)
        axes[row_idx, col_idx].set_title(
            title,
            fontsize=10
        )
        axes[row_idx, col_idx].axis("off")

    axes[row_idx, 0].set_ylabel(
        f"Case {row_idx + 1}\nGT: digit 5",
        fontsize=10
    )

plt.suptitle(
    "Actual Digit 5 Crops Across Input Resolutions",
    fontsize=16,
    y=1.01
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

# The 24 cases where digit 5 was classified as digit 0 at 1280
digit5_errors = full_3res[
    (full_3res["true_class"] == 5) &
    (full_3res["pred_class_1280"] == 0)
].copy()

candidate_results = []

for _, row in digit5_errors.iterrows():

    image = row["image"]
    _, gt_box = get_ground_truth(image, row["gt_index"])

    predictions = pred_groups_960.get(image)

    # Use the 1280 px prediction dataframe instead
    predictions = pred_groups_1280.get(image)

    if predictions is None:
        continue

    for _, pred in predictions.iterrows():

        pred_box = [
            pred["x1"], pred["y1"],
            pred["x2"], pred["y2"]
        ]

        iou = calculate_iou(gt_box, pred_box)

        if iou >= 0.30:
            candidate_results.append({
                "image": image,
                "gt_index": row["gt_index"],
                "predicted_class": int(pred["pred_class"]),
                "confidence": float(pred["confidence"]),
                "IoU": iou
            })

candidates_1280 = pd.DataFrame(candidate_results)

print("OVERLAPPING 1280 PX PREDICTIONS")
print("Number of candidate predictions:", len(candidates_1280))

display(
    candidates_1280.sort_values(
        ["image", "IoU"],
        ascending=[True, False]
    )
)

print("\nSUMMARY BY PREDICTED CLASS")

display(
    candidates_1280.groupby("predicted_class")
    .agg(
        candidate_count=("image", "size"),
        mean_confidence=("confidence", "mean"),
        mean_IoU=("IoU", "mean"),
        max_confidence=("confidence", "max")
    )
    .round(4)
    .sort_values("candidate_count", ascending=False)
)
```

```python
# Reassess the 24 digit-5 cases using all overlapping predictions

candidate_summary = (
    candidates_1280
    .assign(
        is_correct=lambda x: x["predicted_class"] == 5,
        is_digit_zero=lambda x: x["predicted_class"] == 0
    )
    .groupby(["image", "gt_index"])
    .apply(
        lambda g: pd.Series({
            "correct_5_found": g["is_correct"].any(),
            "best_5_confidence": (
                g.loc[g["is_correct"], "confidence"].max()
                if g["is_correct"].any() else 0
            ),
            "best_5_IoU": (
                g.loc[g["is_correct"], "IoU"].max()
                if g["is_correct"].any() else 0
            ),
            "best_0_confidence": (
                g.loc[g["is_digit_zero"], "confidence"].max()
                if g["is_digit_zero"].any() else 0
            ),
            "best_0_IoU": (
                g.loc[g["is_digit_zero"], "IoU"].max()
                if g["is_digit_zero"].any() else 0
            ),
            "highest_confidence": g["confidence"].max(),
            "candidate_count": len(g)
        }),
        include_groups=False
    )
    .reset_index()
)

candidate_summary["correct_5_IoU50"] = (
    candidate_summary["best_5_IoU"] >= 0.50
)

print("PER-CHARACTER REASSESSMENT")
display(candidate_summary)

print("\nSUMMARY")
print("Total ground-truth characters:", len(candidate_summary))
print(
    "Characters with a correct digit-5 candidate:",
    candidate_summary["correct_5_found"].sum()
)
print(
    "Characters with a correct digit-5 candidate at IoU >= 0.50:",
    candidate_summary["correct_5_IoU50"].sum()
)
print(
    "Characters with no correct digit-5 candidate:",
    (~candidate_summary["correct_5_found"]).sum()
)
```

```python
import numpy as np
from scipy.optimize import linear_sum_assignment

# Convert ground-truth YOLO labels into pixel boxes
gt_boxes = []
gt_classes = []

for line in gt_lines:
    values = line.strip().split()
    gt_classes.append(int(values[0]))

    xc, yc, w, h = map(float, values[1:])

    gt_boxes.append([
        (xc - w/2) * width,
        (yc - h/2) * height,
        (xc + w/2) * width,
        (yc + h/2) * height
    ])

gt_boxes = np.array(gt_boxes)
gt_classes = np.array(gt_classes)


def calculate_iou_matrix(gt, pred):
    if len(gt) == 0 or len(pred) == 0:
        return np.zeros((len(gt), len(pred)))

    x1 = np.maximum(gt[:, None, 0], pred[None, :, 0])
    y1 = np.maximum(gt[:, None, 1], pred[None, :, 1])
    x2 = np.minimum(gt[:, None, 2], pred[None, :, 2])
    y2 = np.minimum(gt[:, None, 3], pred[None, :, 3])

    intersection = np.maximum(0, x2-x1) * np.maximum(0, y2-y1)

    gt_area = (
        (gt[:, 2] - gt[:, 0]) *
        (gt[:, 3] - gt[:, 1])
    )

    pred_area = (
        (pred[:, 2] - pred[:, 0]) *
        (pred[:, 3] - pred[:, 1])
    )

    union = gt_area[:, None] + pred_area[None, :] - intersection

    return intersection / np.maximum(union, 1e-10)


def one_to_one_match(gt_boxes, gt_classes, predictions,
                     iou_threshold=0.50, class_aware=False):

    pred_boxes = predictions[
        ["x1", "y1", "x2", "y2"]
    ].to_numpy()

    pred_classes = predictions["pred_class"].to_numpy()

    iou_matrix = calculate_iou_matrix(gt_boxes, pred_boxes)

    valid = iou_matrix >= iou_threshold

    if class_aware:
        valid &= gt_classes[:, None] == pred_classes[None, :]

    # Reward valid matches first, then higher IoU.
    scores = np.where(valid, 1.0 + iou_matrix * 0.001, 0.0)

    # Dummy columns allow GT objects to remain unmatched.
    scores = np.concatenate(
        [scores, np.zeros((len(gt_boxes), len(gt_boxes)))],
        axis=1
    )

    gt_indices, assigned_indices = linear_sum_assignment(
        scores, maximize=True
    )

    matches = []

    for gi, pi in zip(gt_indices, assigned_indices):
        if pi < len(predictions) and valid[gi, pi]:
            matches.append({
                "gt_index": gi,
                "pred_index": pi,
                "iou": iou_matrix[gi, pi],
                "gt_class": gt_classes[gi],
                "pred_class": pred_classes[pi],
                "correct_class": gt_classes[gi] == pred_classes[pi]
            })

    return matches


for resolution, df in [
    (640, pred_640),
    (960, pred_960),
    (1280, pred_1280)
]:
    sample_predictions = df[df["image"] == sample_image].reset_index(drop=True)

    agnostic = one_to_one_match(
        gt_boxes, gt_classes, sample_predictions,
        class_aware=False
    )

    aware = one_to_one_match(
        gt_boxes, gt_classes, sample_predictions,
        class_aware=True
    )

    print(f"\nRESOLUTION: {resolution}")
    print("Ground-truth characters:", len(gt_boxes))
    print("Predictions:", len(sample_predictions))
    print("Class-agnostic matches:", len(agnostic))
    print("Correct classifications among those matches:",
          sum(m["correct_class"] for m in agnostic))
    print("Class-aware matches:", len(aware))

    print("Class-agnostic match details:")
    for m in agnostic:
        print(m)
```

```python
import os
import glob
import numpy as np
import pandas as pd
from PIL import Image
from scipy.optimize import linear_sum_assignment

base = "/content/drive/MyDrive/BLPD/Clean_Test"
analysis_dir = "/content/drive/MyDrive/BLPD/Analysis"
os.makedirs(analysis_dir, exist_ok=True)

prediction_sets = {
    640: pred_640,
    960: pred_960,
    1280: pred_1280
}

# Group predictions by image for faster processing
prediction_groups = {
    resolution: {
        image: group.reset_index(drop=True)
        for image, group in df.groupby("image")
    }
    for resolution, df in prediction_sets.items()
}

image_paths = sorted(
    glob.glob(os.path.join(base, "images", "*.jpg")) +
    glob.glob(os.path.join(base, "images", "*.jpeg")) +
    glob.glob(os.path.join(base, "images", "*.png"))
)

print("Images found:", len(image_paths))


def match_objects(gt_boxes, gt_classes, predictions,
                  iou_threshold=0.50, class_aware=False):

    pred_boxes = predictions[
        ["x1", "y1", "x2", "y2"]
    ].to_numpy()

    pred_classes = predictions["pred_class"].to_numpy()

    if len(gt_boxes) == 0 or len(pred_boxes) == 0:
        return []

    iou_matrix = calculate_iou_matrix(gt_boxes, pred_boxes)

    valid = iou_matrix >= iou_threshold

    if class_aware:
        valid &= gt_classes[:, None] == pred_classes[None, :]

    scores = np.where(valid, 1.0 + iou_matrix * 0.001, 0.0)

    scores = np.concatenate(
        [scores, np.zeros((len(gt_boxes), len(gt_boxes)))],
        axis=1
    )

    gt_indices, assigned_indices = linear_sum_assignment(
        scores, maximize=True
    )

    matches = []

    for gi, pi in zip(gt_indices, assigned_indices):
        if pi < len(predictions) and valid[gi, pi]:
            matches.append({
                "gt_index": int(gi),
                "pred_index": int(pi),
                "iou": float(iou_matrix[gi, pi]),
                "gt_class": int(gt_classes[gi]),
                "pred_class": int(pred_classes[pi]),
                "correct_class": bool(
                    gt_classes[gi] == pred_classes[pi]
                )
            })

    return matches


all_results = []
summary_results = []

for resolution, prediction_map in prediction_groups.items():

    total_gt = 0
    total_predictions = 0
    agnostic_matches = 0
    correct_classifications = 0
    aware_matches = 0

    for image_path in image_paths:

        image_name = os.path.basename(image_path)
        label_path = os.path.join(
            base, "labels",
            os.path.splitext(image_name)[0] + ".txt"
        )

        if not os.path.exists(label_path):
            raise FileNotFoundError(label_path)

        with Image.open(image_path) as img:
            width, height = img.size

        gt_boxes = []
        gt_classes = []
        gt_sizes = []

        with open(label_path, "r") as f:
            for line in f:
                values = line.strip().split()

                if len(values) < 5:
                    continue

                cls = int(values[0])
                xc, yc, w, h = map(float, values[1:5])

                box = [
                    (xc - w/2) * width,
                    (yc - h/2) * height,
                    (xc + w/2) * width,
                    (yc + h/2) * height
                ]

                gt_boxes.append(box)
                gt_classes.append(cls)
                gt_sizes.append(min(box[2] - box[0],
                                    box[3] - box[1]))

        gt_boxes = np.array(gt_boxes, dtype=float).reshape(-1, 4)
        gt_classes = np.array(gt_classes, dtype=int)
        gt_sizes = np.array(gt_sizes, dtype=float)

        predictions = prediction_map.get(
            image_name,
            pd.DataFrame(columns=[
                "image", "x1", "y1", "x2", "y2",
                "confidence", "pred_class"
            ])
        )

        predictions = predictions.reset_index(drop=True)

        agnostic = match_objects(
            gt_boxes, gt_classes, predictions,
            class_aware=False
        )

        aware = match_objects(
            gt_boxes, gt_classes, predictions,
            class_aware=True
        )

        agnostic_by_gt = {
            m["gt_index"]: m for m in agnostic
        }

        aware_gt_indices = {
            m["gt_index"] for m in aware
        }

        total_gt += len(gt_boxes)
        total_predictions += len(predictions)
        agnostic_matches += len(agnostic)
        correct_classifications += sum(
            m["correct_class"] for m in agnostic
        )
        aware_matches += len(aware)

        for gi in range(len(gt_boxes)):

            match = agnostic_by_gt.get(gi)

            all_results.append({
                "resolution": resolution,
                "image": image_name,
                "gt_index": gi,
                "gt_class": int(gt_classes[gi]),
                "gt_min_dimension": float(gt_sizes[gi]),
                "agnostic_matched": match is not None,
                "agnostic_iou": match["iou"] if match else np.nan,
                "pred_class": match["pred_class"] if match else np.nan,
                "pred_confidence": (
                    float(predictions.iloc[match["pred_index"]]["confidence"])
                    if match else np.nan
                ),
                "correct_class": (
                    match["correct_class"] if match else False
                ),
                "class_aware_matched": gi in aware_gt_indices
            })

    summary_results.append({
        "resolution": resolution,
        "ground_truth": total_gt,
        "predictions": total_predictions,
        "agnostic_matches": agnostic_matches,
        "localization_recall_pct": 100 * agnostic_matches / total_gt,
        "correct_classifications": correct_classifications,
        "classification_accuracy_among_matches_pct": (
            100 * correct_classifications / agnostic_matches
            if agnostic_matches else 0
        ),
        "class_aware_TP": aware_matches,
        "class_aware_recall_pct": 100 * aware_matches / total_gt,
        "class_aware_precision_pct": (
            100 * aware_matches / total_predictions
            if total_predictions else 0
        )
    })

    print(
        f"Completed {resolution}: "
        f"{total_gt:,} GT characters, "
        f"{total_predictions:,} predictions"
    )


full_matching_results = pd.DataFrame(all_results)
matching_summary = pd.DataFrame(summary_results)

# Save both detailed and aggregate results
full_matching_results.to_csv(
    os.path.join(analysis_dir, "one_to_one_matching_all_resolutions.csv"),
    index=False
)

matching_summary.to_csv(
    os.path.join(analysis_dir, "one_to_one_matching_summary.csv"),
    index=False
)

print("\nFINAL SUMMARY")
display(matching_summary.round(4))

print("\nDetailed records:", full_matching_results.shape)
print("Saved to:", analysis_dir)
```

```python
import pandas as pd
import numpy as np

df = full_matching_results.copy()

bins = [0, 16, 32, 48, 64, 96, np.inf]
labels = ["<16", "16–31", "32–47", "48–63", "64–95", "≥96"]

df["size_group"] = pd.cut(
    df["gt_min_dimension"],
    bins=bins,
    labels=labels,
    right=False
)

size_results = []

for (resolution, size), group in df.groupby(
    ["resolution", "size_group"],
    observed=False
):

    n = len(group)
    localized = group["agnostic_matched"].sum()

    correct = (
        group.loc[group["agnostic_matched"], "correct_class"].sum()
    )

    class_aware_tp = group["class_aware_matched"].sum()

    size_results.append({
        "resolution": resolution,
        "size_group": str(size),
        "GT_count": n,
        "localized": localized,
        "localization_recall_pct": (
            100 * localized / n if n else np.nan
        ),
        "correct_classifications": correct,
        "classification_accuracy_pct": (
            100 * correct / localized if localized else np.nan
        ),
        "class_aware_TP": class_aware_tp,
        "class_aware_recall_pct": (
            100 * class_aware_tp / n if n else np.nan
        )
    })

size_summary = pd.DataFrame(size_results)

display(
    size_summary.round(2).sort_values(
        ["size_group", "resolution"]
    )
)

size_summary.to_csv(
    "/content/drive/MyDrive/BLPD/Analysis/one_to_one_size_group_analysis.csv",
    index=False
)

print("\nSaved corrected size-group analysis.")
```

```python
import pandas as pd

# Focus on characters with a minimum dimension of at least 32 pixels
class_data = full_matching_results[
    full_matching_results["gt_min_dimension"] >= 32
].copy()

# Aggregate results by resolution and actual character class
class_results = (
    class_data.groupby(["resolution", "gt_class"])
    .agg(
        GT_count=("gt_class", "size"),
        localized=("agnostic_matched", "sum"),
        correct_classifications=("correct_class", "sum"),
        class_aware_TP=("class_aware_matched", "sum")
    )
    .reset_index()
)

# Calculate per-class metrics
class_results["classification_accuracy_pct"] = (
    class_results["correct_classifications"] /
    class_results["localized"].replace(0, pd.NA) * 100
)

class_results["localization_recall_pct"] = (
    class_results["localized"] /
    class_results["GT_count"] * 100
)

class_results["class_aware_recall_pct"] = (
    class_results["class_aware_TP"] /
    class_results["GT_count"] * 100
)

# Retain classes with at least 10 ground-truth instances
# (the GT count is identical across resolutions)
class_support = (
    class_results.groupby("gt_class")["GT_count"].first()
)

eligible_classes = class_support[class_support >= 10].index

class_results_filtered = class_results[
    class_results["gt_class"].isin(eligible_classes)
].copy()

# Compare classification accuracy across resolutions
accuracy_pivot = class_results_filtered.pivot(
    index="gt_class",
    columns="resolution",
    values="classification_accuracy_pct"
)

accuracy_pivot["change_640_to_1280"] = (
    accuracy_pivot[1280] - accuracy_pivot[640]
)

accuracy_pivot = accuracy_pivot.sort_values(
    "change_640_to_1280"
)

# Save complete and filtered results
class_results.to_csv(
    "/content/drive/MyDrive/BLPD/Analysis/one_to_one_per_class_all_sizes32plus.csv",
    index=False
)

accuracy_pivot.to_csv(
    "/content/drive/MyDrive/BLPD/Analysis/per_class_resolution_comparison.csv"
)

print("Total character classes:", len(class_support))
print("Classes with at least 10 instances:", len(eligible_classes))
print("Classes excluded due to low support:",
      len(class_support) - len(eligible_classes))

print("\nLargest classification accuracy declines (640 to 1280):")
display(accuracy_pivot.head(15).round(2))

print("\nLargest classification accuracy improvements (640 to 1280):")
display(accuracy_pivot.tail(15).sort_values(
    "change_640_to_1280", ascending=False
).round(2))
```

```python
# Use the existing per-class comparison table
# and retrieve the ground-truth support for each class.

support = (
    class_results_filtered
    .groupby("gt_class")["GT_count"]
    .first()
)

error_analysis = accuracy_pivot.copy()

error_analysis["GT_count"] = support

# Calculate correct classifications at each resolution
for resolution in [640, 960, 1280]:
    error_analysis[f"errors_{resolution}"] = (
        error_analysis["GT_count"] *
        (1 - error_analysis[resolution] / 100)
    ).round().astype(int)

# Additional errors at 1280 compared with 640
error_analysis["additional_errors_1280_vs_640"] = (
    error_analysis["errors_1280"] -
    error_analysis["errors_640"]
)

error_analysis = error_analysis.sort_values(
    "additional_errors_1280_vs_640",
    ascending=False
)

print("Classes contributing most additional errors at 1280:")
display(error_analysis[
    [
        "GT_count",
        640,
        960,
        1280,
        "errors_640",
        "errors_1280",
        "additional_errors_1280_vs_640"
    ]
].head(20).round(2))

print("\nTotal additional errors across eligible classes:",
      error_analysis["additional_errors_1280_vs_640"].sum())
```

```python
import os
import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import pandas as pd

image_dir = "/content/drive/MyDrive/BLPD/Clean_Test/images"
label_dir = "/content/drive/MyDrive/BLPD/Clean_Test/labels"

pairs_to_inspect = [
    (5, 0),
    (6, 3),
    (38, 59),
    (38, 62)
]

for gt_class, pred_class in pairs_to_inspect:

    subset = full_matching_results[
        (full_matching_results["resolution"] == 1280) &
        (full_matching_results["gt_min_dimension"] >= 32) &
        (full_matching_results["gt_class"] == gt_class) &
        (full_matching_results["pred_class"] == pred_class) &
        (full_matching_results["agnostic_matched"]) &
        (~full_matching_results["correct_class"])
    ].head(3)

    print(f"\n{names[gt_class]} → {names[pred_class]}")

    for _, row in subset.iterrows():

        image_name = row["image"]
        image_path = os.path.join(image_dir, image_name)
        label_path = os.path.join(
            label_dir,
            os.path.splitext(image_name)[0] + ".txt"
        )

        image = cv2.imread(image_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        h, w = image.shape[:2]

        # Retrieve the ground-truth box using gt_index
        labels = pd.read_csv(
            label_path,
            sep=r"\s+",
            header=None
        )

        gt = labels.iloc[int(row["gt_index"])]

        gt_x = float(gt[1]) * w
        gt_y = float(gt[2]) * h
        gt_w = float(gt[3]) * w
        gt_h = float(gt[4]) * h

        gx1 = gt_x - gt_w / 2
        gy1 = gt_y - gt_h / 2

        # Retrieve the prediction box from the 1280 CSV
        pred_row = pred_1280[
            pred_1280["image"] == image_name
        ].copy()

        # Match the same prediction using the stored assignment
        # Re-identify the prediction with maximum IoU and the same class
        # NOTE: this is only a visualization approximation.
        from shapely.geometry import box

        gt_box = box(
            gx1, gy1, gx1 + gt_w, gy1 + gt_h
        )

        pred_row["iou"] = pred_row.apply(
            lambda p: gt_box.intersection(
                box(p.x1, p.y1, p.x2, p.y2)
            ).area / gt_box.union(
                box(p.x1, p.y1, p.x2, p.y2)
            ).area,
            axis=1
        )

        pred_row = pred_row.sort_values("iou", ascending=False).iloc[0]

        fig, ax = plt.subplots(figsize=(9, 6))
        ax.imshow(image)

        # Ground truth: green
        ax.add_patch(patches.Rectangle(
            (gx1, gy1), gt_w, gt_h,
            linewidth=2, edgecolor="lime", facecolor="none",
            label=f"GT: {names[gt_class]}"
        ))

        # Prediction: red
        ax.add_patch(patches.Rectangle(
            (pred_row.x1, pred_row.y1),
            pred_row.x2 - pred_row.x1,
            pred_row.y2 - pred_row.y1,
            linewidth=2, edgecolor="red", facecolor="none",
            label=f"Predicted: {names[pred_class]}"
        ))

        ax.set_title(
            f"{image_name}\n"
            f"IoU: {row['agnostic_iou']:.3f} | "
            f"Confidence: {row['pred_confidence']:.4f}"
        )
        ax.legend()
        ax.axis("off")
        plt.tight_layout()
        plt.show()
```
